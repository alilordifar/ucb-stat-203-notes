# 📊 Conditional Expectation — Complete Notes

> Conditional expectation describes how the **"center"** of one random variable's distribution shifts once we condition on the observed value of another. Just like conditional probability, conditional expectation **shrinks/filters our world** to a specific slice before averaging.

## 1 · Conditional Expectation

**Definition 2.2.10.**
For discrete random variables $X$ and $Y$ with joint PMF $f$:

$$E[Y \mid X = x] = \sum_y y \, f_{Y \mid X}(y \mid x), \quad \forall x \in \text{Supp}[X]$$

For jointly continuous random variables $X$ and $Y$ with joint PDF $f$:

$$E[Y \mid X = x] = \int_{-\infty}^{\infty} y \, f_{Y \mid X}(y \mid x)\, dy, \quad \forall x \in \text{Supp}[X]$$

**🔑 Key insight.** An *unconditional* expectation is a single operator — it takes a random variable and returns one number. A *conditional* expectation is instead a **family of operators**, one for each $x \in \text{Supp}[X]$. Each one summarizes a specific cross-section of the joint distribution: the conditional distribution of $Y$ given $X=x$.

**💡 Tangible example.** Let $X$ be **Bernoulli**, so $\text{Supp}[X]=\{0,1\}$. This gives exactly two operators:

- **$X=0$:** &nbsp; $E[Y\mid X=0]$ → long-run average of $Y$ across all realizations where $X=0$
- **$X=1$:** &nbsp; $E[Y\mid X=1]$ → long-run average of $Y$ across all realizations where $X=1$

## 2 · Conditional Expectation of a Function of Random Variables

**Theorem 2.2.11.**
For discrete $X,Y$ with joint PMF $f$, if $h$ is a function of $X$ and $Y$:

$$E[h(X,Y) \mid X = x] = \sum_y h(x,y)\, f_{Y \mid X}(y \mid x), \quad \forall x \in \text{Supp}[X]$$

For jointly continuous $X,Y$ with joint PDF $f$:

$$E[h(X,Y) \mid X = x] = \int_{-\infty}^{\infty} h(x,y)\, f_{Y \mid X}(y \mid x)\, dy, \quad \forall x \in \text{Supp}[X]$$

## 3 · Conditional Variance

**Definition 2.2.12.**

$$V[Y \mid X = x] = E\big[(Y - E[Y \mid X = x])^2 \mid X = x\big], \quad \forall x \in \text{Supp}[X]$$

**Theorem 2.2.13** *(Alternative formula).* $\forall x \in \text{Supp}[X]$:

$$V[Y \mid X = x] = E[Y^2 \mid X = x] - E[Y \mid X = x]^2$$

## 4 · Linearity of Conditional Expectations

**Theorem 2.2.14.**
For random variables $X,Y$, if $g,h$ are functions of $X$, then $\forall x \in \text{Supp}[X]$:

$$E[g(X)Y + h(X) \mid X = x] = g(x)E[Y \mid X = x] + h(x)$$

<details>
<summary>📝 Proof</summary>

Let $X$ and $Y$ be either discrete with joint PMF $f$ or jointly continuous with joint PDF $f$, and let $g,h$ be functions of $X$.

**Discrete case** — by Theorem 2.2.11, $\forall x \in \text{Supp}[X]$:

$$E[g(X)Y + h(X) \mid X = x] = \sum_y (g(x)y + h(x))\, f_{Y \mid X}(y \mid x)$$

$$= g(x)\sum_y y f_{Y \mid X}(y \mid x) + h(x)\sum_y f_{Y \mid X}(y \mid x)$$

$$= g(x)E[Y \mid X = x] + h(x)\cdot 1 = g(x)E[Y \mid X = x] + h(x)$$

**Continuous case** — by Theorem 2.2.11, $\forall x \in \text{Supp}[X]$:

$$E[g(X)Y + h(X) \mid X = x] = \int_{-\infty}^{\infty} (g(x)y + h(x))\, f_{Y \mid X}(y \mid x)\, dy$$

$$= g(x)\int_{-\infty}^{\infty} y f_{Y \mid X}(y \mid x)\, dy + h(x)\int_{-\infty}^{\infty} f_{Y \mid X}(y \mid x)\, dy$$

$$= g(x)E[Y \mid X = x] + h(x)\cdot 1 = g(x)E[Y \mid X = x] + h(x) \qquad \blacksquare$$

</details>

## 5 · The Conditional Expectation Function (CEF)

The CEF is a single function that takes $x$ as input and returns $E[Y\mid X=x]$ — it characterizes *every* conditional expectation at once.

**Definition 2.2.15.**

$$G_Y(x) = E[Y \mid X = x], \quad \forall x \in \text{Supp}[X]$$

**Notation.** We generally write $E[Y\mid X=x]$ to denote the CEF rather than $G_Y(x)$ — emphasizing the *function* that maps $x \mapsto E[Y\mid X=x]$, not just its value at one point. The CEF is a **univariate function of $x$ only** — it is *not* a function of $Y$. For Bernoulli $X$:

$$
G_Y(x) = E[Y \mid X = x] =
\begin{cases}
E[Y \mid X = 0] & : x = 0 \\
E[Y \mid X = 1] & : x = 1
\end{cases}
$$

Since $G_Y(X)$ is a function of the random variable $X$, it is **itself a random variable** — when $X$ takes value $x$, $G_Y(X)$ takes value $G_Y(x)$. We write $E[Y\mid X]$ to denote $G_Y(X)$ (writing $E[Y\mid X=X]$ would be confusing). For Bernoulli $X$ with $p=\tfrac12$: $E[Y\mid X]$ takes the value $E[Y\mid X=0]$ w.p. $\tfrac12$, and $E[Y\mid X=1]$ w.p. $\tfrac12$.

We can analogously define the **conditional variance function** $H_Y(x) = V[Y\mid X=x]$, generally written $V[Y\mid X=x]$, with $V[Y\mid X]$ denoting the random variable $H_Y(X)$.

By the *Equality of Functions of a Random Variable* theorem, statements about CEFs can be written in the compact $E[Y\mid X]$ form. For example, Theorem 2.2.14 becomes:

$$E[g(X)Y + h(X) \mid X] = g(X)E[Y \mid X] + h(X)$$

From here on, theorems are stated using this more concise notation.

## 6 · Law of Iterated Expectations

Lets us move between conditional and unconditional expectations.

**Theorem 2.2.17.**

$$E[Y] = E[E[Y \mid X]]$$

<details>
<summary>📝 Proof</summary>

**Discrete case:**

$$E[Y] = \sum_y y f_Y(y) = \sum_y y \sum_x f(x,y) = \sum_x \sum_y y f(x,y)$$

$$= \sum_x \sum_y y f_{Y|X}(y|x) f_X(x) = \sum_x \left( \sum_y y f_{Y|X}(y|x) \right) f_X(x)$$

$$= \sum_x E[Y \mid X=x] f_X(x) = E[E[Y \mid X]]$$

**Continuous case:**

$$E[Y] = \int_{-\infty}^{\infty} y f_Y(y)\, dy = \int_{-\infty}^{\infty} y \left( \int_{-\infty}^{\infty} f(x,y)\, dx \right) dy$$

$$= \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} y f_{Y|X}(y|x) f_X(x)\, dy\, dx$$

$$= \int_{-\infty}^{\infty} \left( \int_{-\infty}^{\infty} y f_{Y|X}(y|x)\, dy \right) f_X(x)\, dx$$

$$= \int_{-\infty}^{\infty} E[Y \mid X=x] f_X(x)\, dx = E[E[Y \mid X]] \qquad \blacksquare$$

</details>

**🧭 In plain words.** The unconditional expectation is a **weighted average of conditional expectations**, weighted by the probability distribution of the variable being conditioned on. This directly yields the *Law of Total Variance*, below.

## 7 · Law of Total Variance

**Theorem 2.2.18.**

$$V[Y] = E[V[Y \mid X]] + V[E[Y \mid X]]$$

<details>
<summary>📝 Proof</summary>

$$V[Y] = E[Y^2] - E[Y]^2 = E[E[Y^2 \mid X]] - E[E[Y \mid X]]^2$$

$$= E\big[E[Y^2 \mid X] - E[Y \mid X]^2 + E[Y \mid X]^2\big] - E[E[Y \mid X]]^2$$

$$= E\big[V[Y \mid X] + E[Y \mid X]^2\big] - E[E[Y \mid X]]^2$$

$$= E[V[Y \mid X]] + \Big(E[E[Y \mid X]^2] - E[E[Y \mid X]]^2\Big)$$

$$= E[V[Y \mid X]] + V[E[Y \mid X]] \qquad \blacksquare$$

</details>

**🧭 In plain words.** Total variance in $Y$ decomposes into the average variability **"within"** values of $X$, plus the variability **"across"** values of $X$.

### 💡 Tangible Example

4 people, split into 2 groups ($X$ = gender), height as $Y$:

- **Men:** 175, 185 → mean **180**
- **Women:** 155, 165 → mean **160**
- **Overall mean** (all 4 people): **170**

**🖼️ Visualized**

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/total_variance_viz.png" width="500">
</div>

*Left: raw data — teal arrows are within-group distance, yellow arrows are between-group distance. Right: the same numbers stacked as within (25) + between (100) = total (125).*

**The arithmetic:**

$$E[V[Y\mid X]] = \frac{5^2+5^2+5^2+5^2}{4} = 25 \quad \text{(within-group variance)}$$

$$V[E[Y\mid X]] = \frac{10^2+10^2}{2} = 100 \quad \text{(between-group variance)}$$

$$V[Y] = 25 + 100 = 125$$

**✅ Sanity check** — ignore the groups, take all 4 raw numbers (175, 185, 155, 165), mean = 170:

$$V[Y] = \frac{5^2+15^2+15^2+5^2}{4} = \frac{25+225+225+25}{4} = 125 \quad ✓$$

**🧭 Takeaway.** If you know someone's gender, most of your uncertainty about their height (**100 of 125**) disappears — gender explains where the group sits. The remaining **25** is uncertainty gender *can't* fix: true individual randomness within the group.

## 8 · Properties of Deviations from the CEF

**Theorem 2.2.19.** Let $\varepsilon = Y - E[Y \mid X]$. Then:

- $E[\varepsilon \mid X] = 0$
- $E[\varepsilon] = 0$
- If $g$ is a function of $X$, then $\text{Cov}[g(X), \varepsilon] = 0$
- $V[\varepsilon \mid X] = V[Y \mid X]$
- $V[\varepsilon] = E[V[Y \mid X]]$

<details>
<summary>📝 Proof</summary>

Noting $E[Y\mid X]$ is a function solely of $X$, by Theorem 2.2.14:

$$E[\varepsilon \mid X] = E[Y - E[Y\mid X] \mid X] = E[Y\mid X] - E[Y\mid X] = 0$$

By the Law of Iterated Expectations:

$$E[\varepsilon] = E[E[\varepsilon \mid X]] = E[0] = 0$$

Let $g$ be a function of $X$:

$$\text{Cov}[g(X), \varepsilon] = E[g(X)\varepsilon] - E[g(X)]E[\varepsilon]$$

$$= E\big[g(X)(Y - E[Y\mid X])\big] - E[g(X)](0)$$

$$= E[g(X)Y - g(X)E[Y\mid X]] = E[g(X)Y] - E[g(X)E[Y\mid X]]$$

$$= E[g(X)Y] - E\big[E[g(X)Y \mid X]\big] = E[g(X)Y] - E[g(X)Y] = 0$$

By Definition 2.2.12:

$$V[\varepsilon \mid X] = E\big[(\varepsilon - E[\varepsilon\mid X])^2 \mid X\big] = E[(\varepsilon - 0)^2 \mid X] = E[\varepsilon^2 \mid X]$$

$$= E\big[(Y - E[Y\mid X])^2 \mid X\big] = V[Y\mid X]$$

Finally, by the Law of Total Variance:

$$V[\varepsilon] = E[V[\varepsilon\mid X]] + V[E[\varepsilon\mid X]] = E[V[\varepsilon\mid X]] + V[0]$$

$$= E[V[\varepsilon\mid X]] = E[V[Y\mid X]] \qquad \blacksquare$$

</details>

### 💡 Tangible Example

Reusing the height data — the CEF *is* the group mean, so $\varepsilon$ is the leftover error:

Recall $\varepsilon = Y - E[Y\mid X]$:

| Person | Height (Y) | Prediction | Error (ε) |
|:---|:---:|:---:|:---:|
| Man 1 | 175 | 180 | −5 |
| Man 2 | 185 | 180 | +5 |
| Woman 1 | 155 | 160 | −5 |
| Woman 2 | 165 | 160 | +5 |

**🖼️ Visualized**

</div>

*Left: raw heights with each person's group prediction — yellow arrows are $\varepsilon$. Right: residuals alone — Men's and Women's errors look identical, no group structure survives.*

**The five properties, read off the table:**

- **$E[\varepsilon\mid X]=0$** — within Men: $(-5+5)/2=0$. Within Women: same.
- **$E[\varepsilon]=0$** — average of all four errors is $0$.
- **$\text{Cov}[g(X),\varepsilon]=0$** — the pattern $(-5,+5)$ repeats identically in every group, so $X$ gives zero help predicting $\varepsilon$.
- **$V[\varepsilon\mid X]=V[Y\mid X]$** — variance of $(-5,+5)$ is $25$, matching the within-group variance (subtracting a constant never changes spread).
- **$V[\varepsilon]=E[V[Y\mid X]]=25$** — the between-group variance ($100$) got absorbed into the prediction; only within-group noise ($25$) survives as error.

**🧭 Takeaway.** Once you condition on $X$, the CEF has already extracted everything $X$ can tell you. What's left ($\varepsilon$) is **pure noise**: mean-zero everywhere, uncorrelated with $X$ — exactly why $\varepsilon$ is called a **residual** in regression.
