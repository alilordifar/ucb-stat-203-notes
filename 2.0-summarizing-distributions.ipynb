# Summarizing Distributions

## Probability VS. Statistics
**Probability** says given how data are generated, what is the probability of a particular event? It is forward direction, from model to data. In other words we know the process and we ask what happens? Therefore, going from cause to effect. On the other hand, **statistics** asks what probability distribution could have generated this data? It is backward direction, from data or observations to model. We see data and we ask what caused it? Hence, going from effect to cause.

## Summary Features of Random Variables
We begin by defining measures of the *"center"* and *"spread"* of the distribution of a single random variable.

In the real world, the full probability distribution of a random variable maybe highly complex and therefore may not have a simple mathematical representation. Hence, we will want to be able to describe at least certain key features of such distributions nonparametrically, that is, without assuming that these distributions can be fully characterized by a distribution function with a finite number of parameters.

### Expected Values
The expected value also known as expectation or mean of a random variable can be thought of as the value we would obtain if we took the average over many, many realizations of a random variable. It is the most commonly used measure of the "center" of a probability distribution.

Expected value of a discrete random variable:

$$E[X] = \sum x f(x) $$

Expected value of a continuous random variable:

$$E[x] = \int_{-\infty}^{\infty} x f(x)$$

**Note:** The expected value is an operator which takes on a random variable as an input and returns a number. So, when we compute the expected value of a random variable, we are applying the expectation operator.

---

**Expected Value of Bernoulli Distribution**

$
\mathbb{E}[X] = \sum_{x=0}^{1} x f(x) = 0 \cdot (1 - p) + 1 \cdot p = p
$

**Expected Value of Standard Normal Distribution**

$
\mathbb{E}[X] = \int_{-\infty}^{\infty} x \frac{1}{\sqrt{2\pi}} e^{-x^2/2} \, dx
$

$
= \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} x e^{-x^2/2} \, dx
$

$
= \frac{1}{\sqrt{2\pi}} \left( -e^{-x^2/2} \right)\Big|_{-\infty}^{\infty} = 0
$

---

## 🎲 Law of the Unthinking Statistician (LOTUS)
To find the expected value of a function of a random variable, you don't need the distribution of the function itself. Just use the original variable. In other words, instead of first transforming your data, then figuring out the new distribution, you can skip all that and just apply the function directly inside the expectation. The reason it is called **"unconscious"** is because we don't think about the new distribution at all.

*Want the average of a transformed variable? Just apply the transformation and average using the original probabilities*

**Theorem — Expectation of a Function of a Random Variable (LOTUS)**

**Discrete:**

$
\mathbb{E}[g(X)] = \sum_{x} g(x) f(x)
$

**Continuous:**

$
\mathbb{E}[g(X)] = \int_{-\infty}^{\infty} g(x) f(x)\, dx
$

## 📌 What is LOTUS?

> **LOTUS** lets you compute **E[g(X)]** directly using the distribution of **X** — without ever finding the distribution of **g(X)**.

$$\boxed{E[g(X)] = \sum_x g(x) \cdot P(X = x)}$$

---

## 🧠 Why "Unconscious"?

People use this shortcut *without thinking* — it feels natural to just "plug in" the function.  
LOTUS formally proves **why that instinct is correct**.

---

## ✅ Tangible Example — Casino Die Game

> You roll a fair 6-sided die. You get paid **X² dollars**.

| Roll (x) | Payout g(x) = x² | P(X = x) | Contribution |
|:---:|:---:|:---:|:---:|
| 1 | 1  | 1/6 | 1/6  |
| 2 | 4  | 1/6 | 4/6  |
| 3 | 9  | 1/6 | 9/6  |
| 4 | 16 | 1/6 | 16/6 |
| 5 | 25 | 1/6 | 25/6 |
| 6 | 36 | 1/6 | 36/6 |

$$E[X^2] = \frac{1 + 4 + 9 + 16 + 25 + 36}{6} = \frac{91}{6} \approx \$15.17$$

> 💡 You **never** needed to ask *"What is P(X² = 4)?"* — just use P(X = 2) directly.

## ⚠️ Common Misconception

> *"Does a uniform X mean g(X) is also uniform?"*

**No.** LOTUS says nothing about the *shape* of the transformed distribution.

### Counterexample: X ~ Uniform(0, 1), Y = X²

- X is spread **evenly** over [0, 1]
- Y = X² gets **squished toward 0** — completely non-uniform

$$f_Y(y) = \frac{1}{2\sqrt{y}}, \quad y \in [0, 1] \quad \leftarrow \text{blows up near 0!}$$

---

## 🔑 Key Takeaways

| ✅ What LOTUS *does* say | ❌ What LOTUS *does NOT* say |
|---|---|
| Compute **E[g(X)]** using X's distribution directly | g(X) follows the same distribution type as X |
| Skip finding the full distribution of g(X) | Uniform input → Uniform output |
| Valid for **any** transformation g | The shape of g(X) is preserved |

---

## 🛠️ Why it's Powerful — Variance Shortcut

Since:

$$\text{Var}(X) = E[X^2] - (E[X])^2$$

LOTUS lets you compute **E[X²]** cheaply for any distribution — making variance calculations effortless.

---

## 📐 Summary in One Line

> **LOTUS = computational shortcut for expectations only.**  
> The distribution shape after transformation? That's a separate problem (use change-of-variables).

### Expectation of a Function of Two Random Variables

#### Discrete Case
If \(X\) and \(Y\) are discrete random variables:

$$
\mathbb{E}[h(X, Y)] = \sum_{x} \sum_{y} h(x, y)\, f_{X,Y}(x, y)
$$

where $f_{X,Y}(x, y)$ is the joint probability mass function.

---

#### Continuous Case
If \(X\) and \(Y\) are continuous random variables:

$$
\mathbb{E}[h(X, Y)] = \int \int h(x, y)\, f_{X,Y}(x, y)\, dx\, dy
$$

where $f_{X,Y}(x, y)$ is the joint probability density function.

**Note:** In general, any mathematical expression containing one or more random variables denotes the function of random variables (itself a random variable) defined by that expression. In other words, whatever you do math with random variables, the result is also a random variable. Then, no matter how complicated the math you do to $X$, the answer is still uincertain untill $X$ is actually observed. So the result is itself a new random variable.

**A concrete example:** Say X = tomorrow's temperature (you don't know it yet). If someone asks "what's the temperature in Fahrenheit?" and you calculate (X × 9/5) + 32, that formula is also a random variable — it's uncertain for the same reason X is uncertain.

**The key insight:** randomness "survives" math operations. You can't algebra your way out of uncertainty.

---

### Linearity of Expectation

For any constants $a, b$ and random variables $X, Y$:

$$
\mathbb{E}[aX + bY + c] = a\,\mathbb{E}[X] + b\,\mathbb{E}[Y] + c
$$

**Important:** Linearity holds **even if the variables are dependent**.

### Moments, Variances, and Standard Deviations



**Definition** *$j^{th}$ Raw Moment*

For a random variable $X$ and $j \in \mathbb{N}$, the $j^{th}$ *raw moment* of $X$ is

$$\mu'_j = E\left[X^j\right]$$

**Definition** *$j^{th}$ Central Moment*

For a random variable $X$ and $j \in \mathbb{N}$, the $j^{th}$ *central moment* of $X$ is

$$\mu_j = E\left[(X - E[X])^j\right]$$

Higher moments tell you more about the shape of a distribution beyond just the average:

- j=2 helps you compute variance (spread)
- j=3 helps you compute skewness (how lopsided the distribution is) --> Is distribution tilted left or right?
- j=4 helps you compute kurtosis (how heavy the tails are) --> How heavy the tails are?

In practice, if you fully know all the moments of a distribution, you essentially know the entire distribution's shape — so moments are a compact way to describe and compare probability distributions.

---

#### Skeweness
Because cubing preserves sign:
- Values on the right of mean contribute positive cubed deviations
- Values on the left contribute negative cubed deviations

If the distribution has a long right tail, the big positive deviations dominate → result is positive → right skew.\
If long left tail → big negative deviations dominate → result is negative → left skew.\
If symmetric → they cancel out → result is zero.\
So the odd power acts like a tug of war between left and right. Whichever side wins tells you the skew direction.

---

#### Kurtosis (Power of 4) — Even Power
Now raise deviations to the 4th power and average: E[(X - μ)⁴]\
Because 4th power is always positive, left vs right no longer matters. But here's the key:\
The 4th power brutally amplifies large deviations.
- A deviation of 2 → 2⁴ = 16
- A deviation of 4 → 4⁴ = 256 (8× the value, but 16× the contribution!)

So extreme values far from the mean get massively upweighted. If your distribution has heavy tails (many extreme outliers), those outliers dominate the average and kurtosis is high.\
If values are tightly packed near the mean, deviations are small, 4th powers stay small, kurtosis is low.\
So the even power acts like a spotlight on outliers — it tells you how fat the tails are.

---

### Variance

One of the most frequently employed central moment is the second central moment, also known as the *variance*. Whereas the expected value of a distribution characterizes its location or center. Variance characterizes its variability or spread. Formally, variance measures the expected value of the squared difference between the observed value of X and its mean. Consequently, higher variance implies greater unpredictability.

**Note:** We cannot have negative variance.

**Definition** *Variance*

The *variance* of a random variable $X$ is

$$V[X] = E\left[(X - E[X])^2\right]$$


**Theorem** *Alternative Formula for Variance*

For a random variable $X$,

$$V[X] = E[X^2] - E[X]^2$$

**Proof:**

$V[X] = E\left[(X - E[X])^2\right]$

$= E\left[X^2 - 2XE[X] + E[X]^2\right]$

$= E[X^2] - 2E[XE[X]] + E[E[X]^2]$

$= E[X^2] - 2E[X]E[X] + E[X]^2$

$= E[X^2] - 2E[X]^2 + E[X]^2$

$= E[X^2] - E[X]^2 \quad \square$

**Theorem** *Properties of Variance*

For a random variable $X$,

- $\forall c \in \mathbb{R}, \quad V[X + c] = V[X]$
- $\forall a \in \mathbb{R}, \quad V[aX] = a^2 V[X]$
----

**Proof:** Let $a, c \in \mathbb{R}$. Then

$V[X + c] = E\left[(X + c - E[X + c])^2\right]$

$= E\left[(X + c - E[X] - c)^2\right]$

$= E\left[(X - E[X])^2\right]$

$= V[X],$

and

$V[aX] = E\left[(aX - E[aX])^2\right]$

$= E\left[(aX - aE[X])^2\right]$

$= E\left[a^2(X - E[X])^2\right]$

$= a^2 E\left[(X - E[X])^2\right]$

$= a^2 V[X]. \quad \square$

### Standard Deviation
It is the sqaure root of the variance.

**Definition** *Standard Deviation*

The *standard deviation* of a random variable $X$ is

$$\sigma[X] = \sqrt{V[X]}$$

**Theorem** *Properties of Standard Deviation*

For a random variable $X$,

- $\forall c \in \mathbb{R}, \quad \sigma[X + c] = \sigma[X]$
- $\forall a \in \mathbb{R}, \quad \sigma[aX] = |a|\,\sigma[X]$

**Note:** The *standard deviation* is often preferable to the variance, since it is on the same scale as the random variable of interest.

**Note:** When we scale up the random variable, the standard deviation remains in the same order magnitude as the spread of outcomes. By contrast, the variance of a random variable can "blow up" when we rescale the random variable. Variance is thus more difficult to interpret than standard deviation, since its magnitude does not clearly correspond to the magnitude of the spread of the distribution.

---

#### Chebyshev's Inequality

**Theorem** *Chebyshev's Inequality*

Let $X$ be a random variable with finite $\sigma[X] > 0$. Then, $\forall \epsilon > 0$,

$$\Pr\left[|X - E[X]| \geq \epsilon\sigma[X]\right] \leq \dfrac{1}{\epsilon^2}$$

---

No matter what distribution you have, most of the data must sit close to the mean — and the inequality tells you exactly how much.

**The tangible version:** Imagine you're measuring people's heights. You know the mean and variance, but you don't know the shape of the distribution at all.\
Chebyshev says:\
"At least 75% of people must be within 2 standard deviations of the mean. At least 89% within 3. At least 94% within 4."

**Why is this powerful?** Because it works for any distribution — it doesn't matter if it's normal, skewed, weird-shaped, whatever. You only need to know the mean and variance.
The tradeoff is that the bounds are loose — if you do know the shape (e.g. it's normal), you can make much tighter statements. Chebyshev is the guaranteed worst-case.

- $|X - E[X]|$ — the distance of $X$ from the mean. How far is this outcome from the average?
- $\sigma[X]$ — standard deviation, your unit of "spread".
- $\epsilon$ (epsilon) — a number you choose. It's a multiplier saying "how many standard deviations away?"\
  So $\epsilon\sigma[X]$ means "$\epsilon$ standard deviations away from the mean."
- $\geq \epsilon\sigma[X]$ — the event: $X$ lands at least $\epsilon$ standard deviations away from the mean.
- $\leq \dfrac{1}{\epsilon^2}$ — the probability of that event is at most this small number.

---

**Concrete example:** set $\epsilon = 2$

- $\epsilon\sigma[X]$ = 2 standard deviations away
- Probability of being that far $\leq \dfrac{1}{4}$ = 25%
- So at least 75% of the data is within 2 standard deviations ✓

Set $\epsilon = 3$: probability $\leq \dfrac{1}{9} \approx 11\%$. At least 89% is close. ✓

The beauty is $\epsilon$ is **your choice** — you decide how far "far" means,
and the formula tells you the worst-case probability of being that far.

# From $e$ to the Normal Distribution
### A Step-by-Step Visual Guide

---

## Step 1: What is $e$?

$e \approx 2.718$ is a special mathematical constant. The key thing is not its value — it is what $e^t$ does:

$$e^t \text{ is always strictly positive — it never touches 0, never goes negative}$$

It takes **any real number** as input and maps it to a **strictly positive number**:

$$t \in (-\infty, +\infty) \xrightarrow{\;\;e^t\;\;} \text{output} \in (0, +\infty)$$

| $t$ | $e^t$ |
|:---:|:---:|
| $-\infty$ | $\to 0$ (never reaches it) |
| $-3$ | $0.050$ |
| $-1$ | $0.368$ |
| $0$ | $1$ |
| $1$ | $2.718$ |
| $3$ | $20.09$ |
| $+\infty$ | $\to \infty$ |

**Key rule — the sign of the exponent controls everything:**
- $e^{+\text{something}}$ → output $> 1$, grows toward $\infty$
- $e^{-\text{something}}$ → output $\in (0,\,1)$, shrinks toward $0$

---

## Step 2: $e^x$ and $e^{-x}$

<div align="center">
  
</div>

- Both curves cross at $x = 0$ where $e^0 = 1$
- $e^x$ (blue): shoots up to the right, collapses toward $0$ on the left
- $e^{-x}$ (orange): mirror image — shoots up to the left, collapses on the right
- **Neither is symmetric** — each decays on one side only

---

## Step 3: The Problem — We Need Symmetry

For a bell curve we need a function that:
1. Peaks at the center
2. Decays on **both** sides equally

$e^{-x}$ fails because on the left side ($x < 0$), $-x$ becomes positive, so the curve **grows** instead of decaying.

**The fix:** square $x$ before feeding it into the exponent.

---

## Step 4: $x^2$ and $-x^2$

<div align="center">
  
</div>

**The critical insight — squaring destroys the sign:**

$$(-3)^2 = 9 \qquad \text{and} \qquad (+3)^2 = 9$$

So $x^2$ is **always non-negative** and $-x^2$ is **always non-positive**, regardless of direction. Both sides are now treated identically.

| $x$ | $x^2$ | $-x^2$ |
|:---:|:---:|:---:|
| $0$ | $0$ | $0$ |
| $\pm 1$ | $1$ | $-1$ |
| $\pm 2$ | $4$ | $-4$ |
| $\pm 3$ | $9$ | $-9$ |

---

## Step 5: Plug $-x^2$ into $e$ — The Bell is Born

Now set $t = -x^2$ and feed it into $e^t$:

<div align="center">
  
</div>

**What the exponential does:**

$$x \text{ moves away from } 0 \;\;\to\;\; x^2 \text{ grows} \;\;\to\;\; -x^2 \text{ gets more negative} \;\;\to\;\; e^{-x^2} \text{ shrinks toward } 0$$

| $x$ | $-x^2$ | $e^{-x^2}$ |
|:---:|:---:|:---:|
| $0$ | $0$ | $1.0000$ ← peak |
| $\pm 1$ | $-1$ | $0.3679$ |
| $\pm 2$ | $-4$ | $0.0183$ |
| $\pm 3$ | $-9$ | $0.0001$ |

The exponential does one job: **translate distance from center into height of the curve** — smoothly, symmetrically, always positive.

---

## Step 6: Controlling the Width with $\sigma$

$e^{-x^2}$ is centered at $0$ with a fixed width. To control the spread, divide the exponent by $2\sigma^2$:

$$e^{-\dfrac{x^2}{2\sigma^2}}$$

- Small $\sigma$ → narrow and tall
- Large $\sigma$ → wide and flat
- The $2$ in the denominator is a convention that makes the variance work out to exactly $\sigma^2$

<div align="center">
  
</div>

---

## Step 7: Shifting the Center with $\mu$

To move the peak from $0$ to any value $\mu$, replace $x$ with $(x - \mu)$:

$$e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}$$

- When $x = \mu$: exponent $= 0 \;\to\; e^0 = 1$ ← peak is always at $\mu$
- Moving away from $\mu$ in either direction → exponent grows negative → curve shrinks

<div align="center">
  
</div>
                                                                                           

---

## Step 8: Force Total Area $= 1$

Any PDF must satisfy:

$$\int_{-\infty}^{\infty} f(x)\, dx = 1$$

It can be shown (via a polar coordinates trick) that:

$$\int_{-\infty}^{\infty} e^{-\frac{(x-\mu)^2}{2\sigma^2}}\, dx = \sigma\sqrt{2\pi}$$

So we divide by this constant to normalize:

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \cdot e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}$$

---

## Step 9: The Full Normal Distribution

$$\boxed{f(x) = \frac{1}{\sigma\sqrt{2\pi}}\; e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}, \qquad \forall\, x \in \mathbb{R}}$$

We write $X \sim \mathcal{N}(\mu,\, \sigma^2)$ to say $X$ follows a normal distribution with mean $\mu$ and variance $\sigma^2$.

## Summary: Every Piece Has a Job

| Part               | Job                                                                                      |
| :----------------- | :--------------------------------------------------------------------------------------- |
| (x − μ)²           | Measures squared distance from the mean — always positive, always symmetric              |
| 2σ²                | Controls how fast the tails fall off — larger σ = slower decay = wider bell              |
| e^(−⋯)             | Converts distance into height — far from mean → exponent very negative → height near 0  |
| 1 / (σ√(2π))       | Normalizing constant — ensures total area under the curve = 1                            |

---

## The Full Chain in One Flow

$$x \text{ moves away from } \mu$$

$$\downarrow$$

$$(x - \mu)^2 \text{ grows symmetrically on both sides}$$

$$\downarrow$$

$$-\frac{(x-\mu)^2}{2\sigma^2} \text{ becomes more and more negative}$$

$$\downarrow$$

$$e^{-\frac{(x-\mu)^2}{2\sigma^2}} \text{ shrinks smoothly toward } 0$$

$$\downarrow$$

$$\frac{1}{\sigma\sqrt{2\pi}} \cdot e^{-\frac{(x-\mu)^2}{2\sigma^2}} \text{ integrates to } 1 \;\;\checkmark \;\;\text{— a valid PDF}$$

**"Why does this formula exist?"**

1. I want a curve that is always positive → use $e^t$
2. I want it to peak at the center and decay on both sides → feed it $-x^2$ (squaring kills the sign, negation ensures decay)
3. I want to control where the center is → replace $x$ with $(x - \mu)$
4. I want to control the width → divide by $2\sigma^2$
5. I want total area $= 1$ → divide by $\sigma\sqrt{2\pi}$

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}}\; e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}$$


**Final Definition:**

**Step 1 — Start with the goal**

"The normal distribution is a way to describe how values of a random variable are spread around a mean.
We want a curve that is always positive, symmetric, peaks at the center, and has total area of 1."

---

**Step 2 — Explain the formula is designed, not magic**

"The formula is built from 5 deliberate decisions:"

- We use $e^t$ because it is always positive
- We feed it $-x^2$ so it decays symmetrically on both sides
- We replace $x$ with $(x - \mu)$ to shift the center
- We divide by $2\sigma^2$ to control the width
- We divide by $\sigma\sqrt{2\pi}$ to make total area $= 1$

---

**Step 3 — Explain the parameters**

"It has two parameters: $\mu$ controls where the bell sits, and $\sigma^2$ controls how wide it is.
A large $\sigma^2$ means values are spread far from the mean. A small $\sigma^2$ means they are tightly packed."

---

**Step 4 — One line summary**

"So the normal distribution is just a mathematically precise bell shape —
the formula tells you the probability density at every point $x$."

# Properties of the Normal Distribution
### $X \sim \mathcal{N}(\mu, \sigma^2)$

---

## 1. Mean

$$E[X] = \mu$$

The mean is simply the parameter $\mu$ you put into the distribution. It tells you **where the center of the bell is**.

Two things are true at $\mu$ simultaneously:
- It is the **peak** of the curve (highest point)
- It is the **center of symmetry** (left and right are mirror images)

This is because $(x - \mu)^2$ is minimized at $x = \mu$, making the exponent $0$ and $e^0 = 1$ — the maximum possible output.

<div align="center">
  
</div>

---

## 2. Variance

$$V[X] = \sigma^2$$

The variance is the parameter $\sigma^2$ you put into the distribution. It controls **how spread out** the bell is.

- Large $\sigma^2$ → wide and flat bell → values are far from the mean on average
- Small $\sigma^2$ → narrow and tall bell → values are tightly packed around the mean

<div align="center">
  
</div>

**Why $\sigma^2$ and not $\sigma$?**

Because variance measures average **squared** deviations from the mean:

$$V[X] = E\left[(X - E[X])^2\right] = E\left[(X - \mu)^2\right]$$

Squaring means the unit of variance is the **square** of the unit of $X$.\
If $X$ is in kg, then $V[X]$ is in kg².

---

## 3. Standard Deviation

$$\sigma[X] = \sqrt{V[X]} = \sqrt{\sigma^2} = \sigma$$

The standard deviation is just the square root of the variance. It brings the unit **back to the same unit as $X$**:
- If $X$ is in kg, $\sigma[X]$ is in kg

This is why standard deviation is often preferred in practice — it is directly interpretable on the same scale as your data.

**Geometrically**, $\sigma$ marks the distance from the mean to the **inflection points** of the bell curve — the points where the curve switches from concave to convex:

**In summary:**

| Property | Formula    | What it tells you                              |
| :------- | :--------- | :--------------------------------------------- |
| Mean     | E[X] = μ   | Where the center/peak is                       |
| Variance | V[X] = σ²  | How spread out the values are (in squared units)|
| Std Dev  | σ[X] = σ   | Same as variance but in original units         |

# 📊 Mean Squared Error (MSE)
### Summarizing a Random Variable — and Why the Mean Wins

</div>

---

## 🧭 The Core Question

> **If I must summarize a random variable X with one number c, which c gives me the smallest squared error on average?**

The answer turns out to be:

$$\boxed{c = E[X] \quad \longrightarrow \quad \textbf{The Mean}}$$

---

## 📘 Definition — Mean Squared Error (MSE) about c

For a random variable $X$ and $c \in \mathbb{R}$, the **mean squared error** of $X$ about $c$ is:

$$\text{MSE}(c) = E\left[(X - c)^2\right]$$

> 💡 **Intuition:** We penalize the distance between $X$ and our chosen summary $c$, then average that squared penalty over all outcomes.

---

## 📐 Root Mean Squared Error (RMSE)

$$\text{RMSE}(c) = \sqrt{E\left[(X - c)^2\right]}$$

Much as the **standard deviation** rescales the variance to stay on the same scale as $X$, **RMSE rescales MSE** so that it remains on the same scale as $X - c$.

> For this reason, researchers often prefer to **report RMSE** rather than MSE.

---

## 🔗 Connection to Variance & Raw Moments

| Special Case | Result | Name |
| :----------- | :----- | :--- |
| c = 0        | E[X²]  | 2nd raw moment |
| c = E[X]     | Var(X) | 2nd central moment (variance) |

$$c = E[X] \quad \Rightarrow \quad E\!\left[(X - E[X])^2\right] = \text{Var}(X)$$

---

## 📗 Theorem — Alternative Formula for MSE

For a random variable $X$ and $c \in \mathbb{R}$:

$$E\left[(X-c)^2\right] = \underbrace{V[X]}_{\text{fixed}} + \underbrace{(E[X] - c)^2}_{\text{minimized when } c = E[X]}$$

### 🔍 Proof

We expand and rearrange using the **subtract-and-add** trick (adding and subtracting $E[X]^2$):

$$E\left[(X-c)^2\right] = E\left[X^2 - 2cX + c^2\right]$$

$$= E[X^2] - 2c\,E[X] + c^2 \xrightarrow{\text{subtract/add } E[X]^2}$$

$$= E[X^2] - E[X]^2 + E[X]^2 - 2c\,E[X] + c^2$$

$$= \Big(E[X^2] - E[X]^2\Big) + \Big(E[X]^2 - 2c\,E[X] + c^2\Big)$$

$$= V[X] + \big(E[X] - c\big)^2 \quad \blacksquare$$

---

## 💡 Key Insight — The Bias-Variance Decomposition

The formula $E[(X-c)^2] = V[X] + (E[X]-c)^2$ reveals a beautiful structure:

| Term          | Role                                         | Controllable? |
| :------------ | :------------------------------------------- | :-----------: |
| V[X]          | **Irreducible variance** — inherent randomness of X - how spread/noisy the errors are | ❌ No   |
| (E[X] - c)²   | **Squared bias** — how far our guess c is from the truth - average systematic error | ✅ Yes |

We can only control the **bias term**, and it equals zero exactly when $c = E[X]$.

**Note:** A model can be bad because its prediction is very noisy (high variance), or it is consistently wrog in one direction (high bias). Hence, if a model always predicts 5 units too high, it means the model has low variance and high bias. On the other hand, if predictions jump randomly around truth, there is a high variance and low bias. MSE penalizes both.

## MSE Decomposition

$$\text{MSE} = \text{Var}(e) + (\text{Bias}(e))^2$$

If you are estimating a single random variable $X$ with a constant guess $c$, then:

$$\text{MSE}(c) = E[(X-c)^2] = \text{Var}(X) + (E[X]-c)^2$$

This tells us:

- **Var(X)** = the inherent randomness/noise in the variable (you cannot remove this)
- **(E[X] - c)²** = how far your guess $c$ is from the true mean (your estimation mistake / bias)

So the formula says:

> **Total prediction error = unavoidable randomness + your wrong guessing error**

---

### Example

Suppose the true average of a random variable is $E[X] = 10$, but you decide to always guess $c = 7$.

Then your prediction is systematically wrong by 3 units every time. That part:

$$(E[X] - c)^2 = (10 - 7)^2 = 9$$

is called the **"wrong guessing"** part. It means: even if the data had zero randomness, you would still make an error because your chosen constant $c$ is not centered at the true mean.

---

### Key Result

> **The best constant guess is always the mean.**

Because MSE is minimized when $c = E[X]$. Then the extra error disappears, and only the unavoidable variance remains.

---

## 🏆 Theorem — The Expected Value Minimizes MSE

> For a random variable $X$, the value of $c$ that minimizes the MSE of $X$ about $c$ is $c^* = E[X]$.
>
> → **This is the model that we pick.**

### 🔍 Proof

Using Theorem:

$$\arg\min_{c\,\in\,\mathbb{R}}\; E\left[(X-c)^2\right] = \arg\min_{c\,\in\,\mathbb{R}} \left[V[X] + (E[X]-c)^2\right]$$

$$= \arg\min_{c\,\in\,\mathbb{R}} \;(E[X]-c)^2 \qquad \text{(}V[X]\text{ is fixed w.r.t. }c\text{)}$$

$$= E[X] \quad \blacksquare$$

> **In plain English:** If we had to pick one number as a prediction of the value of $X$, the **"best" choice** (in terms of minimizing MSE) would be $E[X]$.

---

## 📝 Summary

| Concept               | Formula                            | Notes                      |
| :-------------------- | :--------------------------------- | :------------------------- |
| MSE about c           | E[(X - c)²]                        |                            |
| RMSE about c          | √E[(X - c)²]                       | Same scale as X - c        |
| Alternative formula   | E[(X-c)²] = V[X] + (E[X] - c)²    | Bias-Variance split        |
| Optimal predictor     | c* = E[X]                          | Minimizes MSE              |
| Minimum MSE           | min E[(X-c)²] = V[X]               | Irreducible floor          |
| Special case c = 0    | E[X²] = 2nd raw moment             |                            |
| Special case c = E[X] | Var(X) = 2nd central moment        |                            |

**FINAL MSE Intuition:** Under MSE, the best prediction is the number that balances all squared distances to the data. That balancing point turns out to be the mean.

## Summary: Features of Joint Distributions

> Our goal is to describe the **relationship between random variables**. For example: if one quantity $X$ is high, is another $Y$ more or less likely to be high?

---

## 📐 Covariance

**Covariance** measures the extent to which two random variables *move together*.

$$\text{Cov}[X, Y] = E\big[(X - E[X])(Y - E[Y])\big]$$

| Sign of Cov[X, Y] | Meaning |
| :--- | :--- |
| Positive | X and Y tend to be large (or small) together |
| Negative | When X is large, Y tends to be small (and vice versa) |
| Zero | No linear relationship |

> **Note:** Cov[·, ·] is an operator, not a function of the values its arguments take on — so Cov[X, Y] is a **constant**, not a random variable.

---

## 📗 Theorem — Alternative Formula for Covariance

For random variables $X$ and $Y$:

$$\boxed{\text{Cov}[X, Y] = E[XY] - E[X]\,E[Y]}$$

### 🔍 Proof

$$\text{Cov}[X, Y] = E\big[(X - E[X])(Y - E[Y])\big]$$

$$= E\big[XY - XE[Y] - YE[X] + E[X]E[Y]\big]$$

$$= E[XY] - E[X]E[Y] - E[X]E[Y] + E[X]E[Y]$$

$$= E[XY] - E[X]E[Y] \quad \blacksquare$$

---

## 📗 Theorem — Variance Rule → *Variance of Sums*

Let $X$ and $Y$ be random variables. Then:

$$\boxed{V[X + Y] = V[X] + 2\,\text{Cov}[X, Y] + V[Y]}$$

More generally, $\forall\, a, b, c \in \mathbb{R}$:

$$\boxed{V[aX + bY + c] = a^2 V[X] + 2ab\,\text{Cov}[X, Y] + b^2 V[Y]}$$

### 🔍 Proof

$$V[X + Y] = E[(X+Y)^2] - E[X+Y]^2$$

$$= E[X^2 + 2XY + Y^2] - (E[X] + E[Y])^2$$

$$= E[X^2] + 2E[XY] + E[Y^2] - E[X]^2 - 2E[X]E[Y] - E[Y]^2$$

$$= \big(E[X^2] - E[X]^2\big) + 2\big(E[XY] - E[X]E[Y]\big) + \big(E[Y^2] - E[Y]^2\big)$$

$$= V[X] + 2\,\text{Cov}[X, Y] + V[Y] \quad \blacksquare$$

---

## 📗 Theorem — Properties of Covariance

For random variables $X$, $Y$, $Z$, and $W$, and constants $a, b, c, d \in \mathbb{R}$:

- **Constants have zero covariance:** Cov[c, X] = Cov[X, c] = Cov[c, d] = 0
- **Symmetry:** Cov[X, Y] = Cov[Y, X]
- **Self-covariance is variance:** Cov[X, X] = V[X]
- **Linearity:** Cov[aX + c, bY + d] = ab Cov[X, Y]
- **Bilinearity:** Cov[X + W, Y + Z] = Cov[X, Y] + Cov[X, Z] + Cov[W, Y] + Cov[W, Z]

### 🔍 Proof

**Constants have zero covariance:**

$$\text{Cov}[c, X] = E[cX] - E[c]E[X] = cE[X] - cE[X] = 0$$

$$\text{Cov}[c, d] = E[cd] - E[c]E[d] = cd - cd = 0$$

**Symmetry:**

$$\text{Cov}[X, Y] = E[XY] - E[X]E[Y] = E[YX] - E[Y]E[X] = \text{Cov}[Y, X]$$

**Self-covariance is variance:**

$$\text{Cov}[X, X] = E[XX] - E[X]E[X] = E[X^2] - E[X]^2 = V[X]$$

**Linearity:**

$$\text{Cov}[aX + c,\; bY + d] = E\big[(aX + c - E[aX+c])(bY + d - E[bY+d])\big]$$

$$= E\big[(aX + c - aE[X] - c)(bY + d - bE[Y] - d)\big]$$

$$= E\big[a(X - E[X]) \cdot b(Y - E[Y])\big]$$

$$= ab\, E\big[(X - E[X])(Y - E[Y])\big] = ab\,\text{Cov}[X, Y]$$

**Bilinearity:**

$$\text{Cov}[X+W,\; Y+Z] = E[(X+W)(Y+Z)] - E[X+W]E[Y+Z]$$

$$= E[XY + XZ + WY + WZ] - (E[X]+E[W])(E[Y]+E[Z])$$

$$= \big(E[XY] - E[X]E[Y]\big) + \big(E[XZ] - E[X]E[Z]\big)$$
$$\quad + \big(E[WY] - E[W]E[Y]\big) + \big(E[WZ] - E[W]E[Z]\big)$$

$$= \text{Cov}[X,Y] + \text{Cov}[X,Z] + \text{Cov}[W,Y] + \text{Cov}[W,Z] \quad \blacksquare$$

> 💡 **Key insight:** The third property, Cov[X, X] = V[X], tells us that **variance is a special case of covariance** — the covariance of a random variable with itself. Covariance is a natural generalization of variance.
---

## Covariance — Intuition with Real Data

> Covariance measures whether two variables move together relative to their means.

$$\text{Cov}(X,Y) = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})$$

---

### Real Example — Experience vs Salary

| Person | Experience (X) | Salary (Y) |
| :----- | :------------- | :--------- |
| A      | 1              | 50         |
| B      | 2              | 60         |
| C      | 3              | 70         |
| D      | 4              | 80         |

---

### Step 1 — Compute Means

$$\bar{X} = 2.5 \qquad \bar{Y} = 65$$

---

### Step 2 — Measure Deviations from Mean

| Person | xi - X̄ | yi - Ȳ | Product |
| :----- | :------ | :----- | :------ |
| A      | -1.5    | -15    | +22.5   |
| B      | -0.5    | -5     | +2.5    |
| C      | +0.5    | +5     | +2.5    |
| D      | +1.5    | +15    | +22.5   |

$$\text{Cov}(X,Y) = \frac{22.5 + 2.5 + 2.5 + 22.5}{4} = +12.5$$

---

### What is Happening Here?

- Low experience → low salary → negative × negative = **positive**
- High experience → high salary → positive × positive = **positive**

Every row contributes a positive value, so the covariance is strongly positive.

---

### The Real Intuition

> "When one variable moves away from its mean, does the other move in the **same direction**?"

| Case | Sign pattern | Product | Covariance direction |
| :--- | :----------- | :------ | :------------------- |
| Both above mean | (+)(+) | positive | pushes Cov up |
| Both below mean | (−)(−) | positive | pushes Cov up |
| Opposite sides  | (+)(−) or (−)(+) | negative | pulls Cov down |

---

### Positive Covariance

Both variables move together.

- Experience ↑ → Salary ↑
- Study hours ↑ → Exam score ↑

---

### Negative Covariance

Variables move in opposite directions.

- Speed ↑ → Travel time ↓
- Price ↑ → Demand ↓

---

### Near-Zero Covariance

No consistent relationship — positive and negative products cancel each other out, leaving a sum close to zero.

---

### Why Perfect Lines Give Strong Covariance

If data lies perfectly on a line, whenever X is above its mean Y is also above its mean, and vice versa. Every product has the same sign, so they all reinforce each other rather than cancelling — producing the largest possible covariance.

---

## 📘 Definition — Correlation

> **Correlation is a rescaled version of covariance that captures the linear dependence between two random variables.** Importantly, since the denominator of the expression for the correlation is positive, correlation is positive when covariance is positive and negative when covariance is negative.

The **correlation** of two random variables $X$ and $Y$ with $\sigma[X] > 0$ and $\sigma[Y] > 0$ is:

$$\boxed{\rho[X, Y] = \frac{\text{Cov}[X, Y]}{\sigma[X]\,\sigma[Y]}}$$

| Value of rho | Meaning |
| :--- | :--- |
| rho = 1 | Perfect positive linear relationship |
| rho = -1 | Perfect negative linear relationship |
| rho = 0 | No linear correlation |

> Correlation is bounded: $-1 \leq \rho[X, Y] \leq 1$, making it easier to interpret than raw covariance which has no fixed scale.

### From Covariance to Correlation

Covariance still depends on the **scale and units** of the variables. For example, pairing salary in dollars with height in centimeters produces a huge covariance simply because the numbers are large — not because the relationship is strong.

So covariance alone is hard to interpret across different contexts.

---

#### Correlation Fixes This

Correlation divides by the typical spread of both variables, which rescales the result to always land between −1 and +1, regardless of units.

---

> **Important Note About Correlation**
>
> Correlation does **not** tell us how much one variable numerically changes another variable.  
> For example, a correlation of \(0.8\) does **not** mean that a 1-unit increase in \(X\) causes an 80% increase in \(Y\). Instead, correlation measures how strongly and consistently two variables move together in a linear way.
>
> A high positive correlation means:
>
> - when \(X\) is above its mean, \(Y\) also tends to be above its mean
> - when \(X\) is below its mean, \(Y\) also tends to be below its mean
>
> In other words, correlation measures **synchronization of movement**, not the amount of change.
>
> The quantity that measures “how much \(Y\) changes when \(X\) changes by 1 unit” is called the **slope** in regression, not correlation.

---

#### The Tangible Intuition

- **Covariance asks:** "Do they move together?"
- **Std deviation says:** "How much does each variable naturally fluctuate on its own?"
- **Correlation asks:** "How strong is the co-movement *relative to* their own natural variability?"

$$\rho(X,Y) = \frac{\text{co-movement}}{\text{typical spread of X} \times \text{typical spread of Y}}$$
