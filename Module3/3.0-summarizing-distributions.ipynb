## Probability vs. Statistics

**Probability** says: given how data are generated, what is the probability of a particular event? It is the *forward* direction — from model to data. We know the process and ask what happens. Going from **cause to effect**.

**Statistics** asks: what probability distribution could have generated this data? It is the *backward* direction — from data to model. We see data and ask what caused it. Going from **effect to cause**.

---

## Summary Features of Random Variables

We begin by defining measures of the *"center"* and *"spread"* of the distribution of a single random variable.

In the real world, the full probability distribution of a random variable may be highly complex and may not have a simple mathematical representation. Hence, we want to describe key features of such distributions **nonparametrically** — that is, without assuming these distributions can be fully characterized by a function with a finite number of parameters.

---

## Expected Values

The **expected value** (also known as *expectation* or *mean*) of a random variable can be thought of as the value we would obtain if we took the average over many, many realizations. It is the most commonly used measure of the "center" of a probability distribution.

**Expected value of a discrete random variable:**

$$E[X] = \sum_x x\, f(x)$$

**Expected value of a continuous random variable:**

$$E[X] = \int_{-\infty}^{\infty} x\, f(x)\, dx$$

> **Note:** The expected value is an **operator** — it takes a random variable as input and returns a number.

---

### Expected Value of the Bernoulli Distribution

$$\mathbb{E}[X] = \sum_{x=0}^{1} x\, f(x) = 0 \cdot (1 - p) + 1 \cdot p = p$$

---

### Expected Value of the Standard Normal Distribution

$$\mathbb{E}[X] = \int_{-\infty}^{\infty} x \frac{1}{\sqrt{2\pi}} e^{-x^2/2}\, dx$$

$$= \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} x\, e^{-x^2/2}\, dx$$

$$= \frac{1}{\sqrt{2\pi}} \left( -e^{-x^2/2} \right)\Big|_{-\infty}^{\infty} = 0$$

---

## Law of the Unconscious Statistician (LOTUS)

To find the expected value of a function of a random variable, you **don't need the distribution of the function itself** — just use the original variable. Instead of first transforming your data and then figuring out the new distribution, you can skip all that and apply the function directly inside the expectation.

> The reason it is called **"unconscious"** is because we don't think about the new distribution at all.
> *Want the average of a transformed variable? Just apply the transformation and average using the original probabilities.*

### Theorem — Expectation of a Function of a Random Variable (LOTUS)

**Discrete:**

$$\mathbb{E}[g(X)] = \sum_{x} g(x)\, f(x)$$

**Continuous:**

$$\mathbb{E}[g(X)] = \int_{-\infty}^{\infty} g(x)\, f(x)\, dx$$

---

### What is LOTUS?

> **LOTUS** lets you compute $E[g(X)]$ directly using the distribution of $X$ — without ever finding the distribution of $g(X)$.

$$\boxed{E[g(X)] = \sum_x g(x) \cdot P(X = x)}$$

---

### Why "Unconscious"?

People use this shortcut *without thinking* — it feels natural to just "plug in" the function.
LOTUS formally proves **why that instinct is correct**.

---

### Tangible Example — Casino Die Game

> You roll a fair 6-sided die. You get paid **X² dollars**.

| Roll x | Payout x² | P(X=x) | Contribution |
|:---:|:---:|:---:|:---:|
| 1 | 1  | 1/6 | 1/6  |
| 2 | 4  | 1/6 | 4/6  |
| 3 | 9  | 1/6 | 9/6  |
| 4 | 16 | 1/6 | 16/6 |
| 5 | 25 | 1/6 | 25/6 |
| 6 | 36 | 1/6 | 36/6 |

$$E[X^2] = \frac{1 + 4 + 9 + 16 + 25 + 36}{6} = \frac{91}{6} \approx \$15.17$$

> 💡 You **never** needed to ask *"What is P(X² = 4)?"* — just use P(X = 2) directly.

---

### ⚠️ Common Misconception

> *"Does a uniform X mean g(X) is also uniform?"*

**No.** LOTUS says nothing about the *shape* of the transformed distribution.

**Counterexample: X ~ Uniform(0, 1), Y = X²**

- X is spread **evenly** over [0, 1]
- Y = X² gets **squished toward 0** — completely non-uniform

$$f_Y(y) = \frac{1}{2\sqrt{y}}, \quad y \in [0, 1] \quad \leftarrow \text{blows up near 0!}$$

---

### Key Takeaways

| LOTUS does say | LOTUS does NOT say |
|---|---|
| Use X's distribution to find E[g(X)] | g(X) has same distribution as X |
| Skip finding the distribution of g(X) | Uniform input → Uniform output |
| Valid for any transformation g | The shape of g(X) is preserved |

---

### Why LOTUS is Powerful — Variance Shortcut

Since:

$$\text{Var}(X) = E[X^2] - (E[X])^2$$

LOTUS lets you compute $E[X^2]$ cheaply for any distribution — making variance calculations effortless.

> **LOTUS = computational shortcut for expectations only.**
> The distribution shape after transformation? That's a separate problem (use change-of-variables).

## Expectation of a Function of Two Random Variables

### Discrete Case

If $X$ and $Y$ are discrete random variables:

$$\mathbb{E}[h(X, Y)] = \sum_{x} \sum_{y} h(x, y)\, f_{X,Y}(x, y)$$

where $f_{X,Y}(x, y)$ is the joint probability mass function.

### Continuous Case

If $X$ and $Y$ are continuous random variables:

$$\mathbb{E}[h(X, Y)] = \int \int h(x, y)\, f_{X,Y}(x, y)\, dx\, dy$$

where $f_{X,Y}(x, y)$ is the joint probability density function.

> **Note:** Any mathematical expression containing one or more random variables is itself a random variable. No matter how complicated the math you do to $X$, the result is still uncertain until $X$ is actually observed — **randomness "survives" math operations. You can't algebra your way out of uncertainty.**

**Concrete example:** Say $X$ = tomorrow's temperature (unknown). If you calculate $(X \times 9/5) + 32$ to convert to Fahrenheit, that formula is also a random variable — uncertain for the same reason $X$ is.

---

## Linearity of Expectation

For any constants $a, b, c$ and random variables $X, Y$:

$$\mathbb{E}[aX + bY + c] = a\,\mathbb{E}[X] + b\,\mathbb{E}[Y] + c$$

> ⚠️ **Important:** Linearity holds **even if the variables are dependent**.

## Moments, Variances, and Standard Deviations

### Definition — j-th Raw Moment

For a random variable $X$ and $j \in \mathbb{N}$, the $j^{th}$ *raw moment* of $X$ is:

$$\mu'_j = E\left[X^j\right]$$

### Definition — j-th Central Moment

For a random variable $X$ and $j \in \mathbb{N}$, the $j^{th}$ *central moment* of $X$ is:

$$\mu_j = E\left[(X - E[X])^j\right]$$

Higher moments tell you more about the shape of a distribution beyond just the average:

| Moment | Computes | Tells you |
|--------|----------|-----------|
| j = 2 | Variance | Spread around the mean |
| j = 3 | Skewness | Left or right tilt? |
| j = 4 | Kurtosis | How heavy are the tails? |

> In practice, if you fully know all the moments of a distribution, you essentially know the entire distribution's shape — moments are a compact way to describe and compare probability distributions.

---

### Skewness (3rd Moment — Odd Power)

Because cubing preserves sign:
- Values **right** of the mean contribute **positive** cubed deviations
- Values **left** of the mean contribute **negative** cubed deviations

- Long right tail → big positive deviations dominate → **positive skew**
- Long left tail → big negative deviations dominate → **negative skew**
- Symmetric → they cancel out → **zero skew**

> The odd power acts like a tug of war between left and right — whichever side wins tells you the skew direction.

---

### Kurtosis (4th Moment — Even Power)

Raise deviations to the 4th power: $E[(X - \mu)^4]$

Because the 4th power is always positive, left vs. right no longer matters. But the 4th power **brutally amplifies large deviations**:

- A deviation of 2 → $2^4 = 16$
- A deviation of 4 → $4^4 = 256$ ← 16× the contribution!

So extreme values far from the mean get massively upweighted:
- **Heavy tails** (many extreme outliers) → high kurtosis
- **Values tightly packed near the mean** → low kurtosis

> The even power acts like a spotlight on outliers — it tells you how fat the tails are.

---

## Variance

One of the most frequently employed central moments is the **second central moment**, also known as the *variance*. Whereas the expected value characterizes the *location* of a distribution, variance characterizes its *variability* or spread — the expected squared difference between the observed value and the mean.

> **Note:** Variance cannot be negative.

### Definition — Variance

$$V[X] = E\left[(X - E[X])^2\right]$$

### Theorem — Alternative Formula for Variance

$$V[X] = E[X^2] - E[X]^2$$

**Proof:**

$$V[X] = E\left[(X - E[X])^2\right]$$

$$= E\left[X^2 - 2X\,E[X] + E[X]^2\right]$$

$$= E[X^2] - 2E[X]E[X] + E[X]^2$$

$$= E[X^2] - 2E[X]^2 + E[X]^2$$

$$= E[X^2] - E[X]^2 \quad \square$$

---

### Theorem — Properties of Variance

For a random variable $X$:

- $\forall c \in \mathbb{R}, \quad V[X + c] = V[X]$
- $\forall a \in \mathbb{R}, \quad V[aX] = a^2\, V[X]$

**Proof:** Let $a, c \in \mathbb{R}$. Then:

$$V[X + c] = E\left[(X + c - E[X + c])^2\right] = E\left[(X + c - E[X] - c)^2\right] = E\left[(X - E[X])^2\right] = V[X]$$

$$V[aX] = E\left[(aX - E[aX])^2\right] = E\left[(aX - aE[X])^2\right] = E\left[a^2(X - E[X])^2\right] = a^2\, V[X] \quad \square$$

## Standard Deviation

### Definition — Standard Deviation

$$\sigma[X] = \sqrt{V[X]}$$

### Theorem — Properties of Standard Deviation

For a random variable $X$:

- $\forall c \in \mathbb{R}, \quad \sigma[X + c] = \sigma[X]$
- $\forall a \in \mathbb{R}, \quad \sigma[aX] = |a|\,\sigma[X]$

> **Note:** The standard deviation is often preferable to the variance since it is on the **same scale** as the random variable of interest.

> **Note:** When we scale up the random variable, the standard deviation remains in the same order of magnitude as the spread of outcomes. By contrast, the variance can "blow up" when we rescale — making it harder to interpret, since its magnitude doesn't clearly correspond to the magnitude of the spread.

---

## Chebyshev's Inequality

### Theorem — Chebyshev's Inequality

Let $X$ be a random variable with finite $\sigma[X] > 0$. Then $\forall\, \epsilon > 0$:

$$\Pr\left[|X - E[X]| \geq \epsilon\,\sigma[X]\right] \leq \dfrac{1}{\epsilon^2}$$

No matter what distribution you have, most of the data must sit close to the mean — and the inequality tells you exactly how much.

> **Tangible version:** Imagine measuring people's heights. You know the mean and variance, but not the shape of the distribution.
> Chebyshev says: at least 75% of people must be within 2 standard deviations of the mean. At least 89% within 3. At least 94% within 4.

**Why is this powerful?** It works for *any* distribution — normal, skewed, or weird-shaped. You only need the mean and variance. The tradeoff is that bounds are loose — if you know the shape, you can make much tighter statements. Chebyshev is the guaranteed **worst-case**.

**Breaking down the formula:**

| Term | Meaning |
|------|---------|
| \|X − E[X]\| | Distance of X from the mean |
| σ[X] | Standard deviation — unit of spread |
| ε | Multiplier: how many std devs away? |
| ε·σ[X] | ε standard deviations from mean |
| ≤ 1/ε² | Worst-case probability bound |

**Concrete examples:**

- Set $\epsilon = 2$: probability of being $\geq 2\sigma$ away $\leq \frac{1}{4} = 25\%$ → at least **75%** is within 2 std devs ✓
- Set $\epsilon = 3$: probability $\leq \frac{1}{9} \approx 11\%$ → at least **89%** is close ✓

> The beauty is $\epsilon$ is **your choice** — you decide how far "far" means, and the formula gives you the worst-case probability.

## Normal Distribution — From $e$ to the Formula

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/exponential.png" width="400">
</div>

### Step 1: What is $e$?

$e \approx 2.718$ is a special mathematical constant. The key property: $e^t$ is **always strictly positive** — it never touches 0, never goes negative.

$$t \in (-\infty, +\infty) \xrightarrow{\;\;e^t\;\;} \text{output} \in (0, +\infty)$$

| t | e^t |
|:---:|:---:|
| −∞ | → 0 (never reaches it) |
| −3 | 0.050 |
| −1 | 0.368 |
| 0 | 1 |
| 1 | 2.718 |
| 3 | 20.09 |
| +∞ | → ∞ |

**Key rule:** the sign of the exponent controls everything:
- $e^{+\text{something}}$ → output $> 1$, grows toward $\infty$
- $e^{-\text{something}}$ → output $\in (0,\,1)$, shrinks toward $0$

---

### Step 2: $e^x$ and $e^{-x}$

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot1_ex_enx.png" width="400">
</div>

- Both curves cross at $x = 0$ where $e^0 = 1$
- $e^x$ (blue): shoots up to the right, collapses toward $0$ on the left
- $e^{-x}$ (orange): mirror image — shoots up to the left, collapses on the right
- **Neither is symmetric** — each decays on one side only

---

### Step 3: The Problem — We Need Symmetry

For a bell curve we need a function that:
1. Peaks at the center
2. Decays on **both** sides equally

$e^{-x}$ fails because on the left side ($x < 0$), $-x$ becomes positive, so the curve **grows** instead of decaying.

**The fix:** square $x$ before feeding it into the exponent.

---

### Step 4: $x^2$ and $-x^2$

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot2_x2.png" width="400">
</div>

**The critical insight — squaring destroys the sign:**

$$(-3)^2 = 9 \qquad \text{and} \qquad (+3)^2 = 9$$

So $x^2$ is **always non-negative** and $-x^2$ is **always non-positive**, regardless of direction.

| x | x² | −x² |
|:---:|:---:|:---:|
| 0 | 0 | 0 |
| ±1 | 1 | −1 |
| ±2 | 4 | −4 |
| ±3 | 9 | −9 |

---

### Step 5: Plug $-x^2$ into $e$ — The Bell is Born

Now set $t = -x^2$ and feed it into $e^t$:

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot3_chain.png" width="400">
</div>

$$x \text{ moves away from } 0 \;\to\; x^2 \text{ grows} \;\to\; -x^2 \text{ gets more negative} \;\to\; e^{-x^2} \text{ shrinks toward } 0$$

| x | −x² | e^(−x²) |
|:---:|:---:|:---:|
| 0 | 0 | 1.0000 ← peak |
| ±1 | −1 | 0.3679 |
| ±2 | −4 | 0.0183 |
| ±3 | −9 | 0.0001 |

The exponential does one job: **translate distance from center into height of the curve** — smoothly, symmetrically, always positive.

---

### Step 6: Controlling the Width with $\sigma$

$e^{-x^2}$ is centered at $0$ with a fixed width. To control spread, divide the exponent by $2\sigma^2$:

$$e^{-\dfrac{x^2}{2\sigma^2}}$$

- Small $\sigma$ → narrow and tall
- Large $\sigma$ → wide and flat
- The $2$ in the denominator is a convention that makes the variance work out to exactly $\sigma^2$

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot4_sigma.png" width="400">
</div>

---

### Step 7: Shifting the Center with $\mu$

To move the peak from $0$ to any value $\mu$, replace $x$ with $(x - \mu)$:

$$e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}$$

- When $x = \mu$: exponent $= 0 \;\to\; e^0 = 1$ ← peak is always at $\mu$
- Moving away from $\mu$ in either direction → exponent grows negative → curve shrinks

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot5_mu.png" width="400">
</div>

---

### Step 8: Force Total Area $= 1$

Any PDF must satisfy:

$$\int_{-\infty}^{\infty} f(x)\, dx = 1$$

It can be shown (via a polar coordinates trick) that:

$$\int_{-\infty}^{\infty} e^{-\frac{(x-\mu)^2}{2\sigma^2}}\, dx = \sigma\sqrt{2\pi}$$

So we divide by this constant to normalize:

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \cdot e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}$$

---

### Step 9: The Full Normal Distribution

$$\boxed{f(x) = \frac{1}{\sigma\sqrt{2\pi}}\; e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}, \qquad \forall\, x \in \mathbb{R}}$$

We write $X \sim \mathcal{N}(\mu,\, \sigma^2)$ to say $X$ follows a normal distribution with mean $\mu$ and variance $\sigma^2$.

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/plot6_normal.png" width="400">
</div>

---

### Summary: Every Piece Has a Job

| Part | Job |
|------|-----|
| (x − μ)² | Squared distance from mean — always positive, symmetric |
| 2σ² | Controls tail decay — larger σ = wider bell |
| e^(−···) | Converts distance to height — far from mean → near 0 |
| 1 / (σ√2π) | Normalizing constant — total area = 1 |

---

### The Full Chain in One Flow

$$x \text{ moves away from } \mu$$
$$\downarrow$$
$$(x - \mu)^2 \text{ grows symmetrically on both sides}$$
$$\downarrow$$
$$-\frac{(x-\mu)^2}{2\sigma^2} \text{ becomes more and more negative}$$
$$\downarrow$$
$$e^{-\frac{(x-\mu)^2}{2\sigma^2}} \text{ shrinks smoothly toward } 0$$
$$\downarrow$$
$$\frac{1}{\sigma\sqrt{2\pi}} \cdot e^{-\frac{(x-\mu)^2}{2\sigma^2}} \text{ integrates to } 1 \;\checkmark \;\text{— a valid PDF}$$

**"Why does this formula exist?"** — Five deliberate decisions:

1. I want a curve that is always positive → use $e^t$
2. I want it to peak at the center and decay on both sides → feed it $-x^2$ (squaring kills the sign, negation ensures decay)
3. I want to control where the center is → replace $x$ with $(x - \mu)$
4. I want to control the width → divide by $2\sigma^2$
5. I want total area $= 1$ → divide by $\sigma\sqrt{2\pi}$

$$\boxed{f(x) = \frac{1}{\sigma\sqrt{2\pi}}\; e^{-\dfrac{(x-\mu)^2}{2\sigma^2}}}$$

## Properties of the Normal Distribution

### $X \sim \mathcal{N}(\mu, \sigma^2)$

---

### 1. Mean

$$E[X] = \mu$$

The mean is the parameter $\mu$. Two things are true at $\mu$ simultaneously:
- It is the **peak** of the curve (highest point)
- It is the **center of symmetry** (left and right are mirror images)

This is because $(x - \mu)^2$ is minimized at $x = \mu$, making the exponent $0$ and $e^0 = 1$ — the maximum output.

  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/prop1_mean.png" width="400">
</div>

---

### 2. Variance

$$V[X] = \sigma^2$$

| σ² | Effect |
|----|--------|
| Large | Wide flat bell — values far from mean |
| Small | Narrow tall bell — values near the mean |

<div align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/prop2_variance.png" width="400">
</div>

**Why σ² and not σ?**

Because variance measures average **squared** deviations:

$$V[X] = E\left[(X - E[X])^2\right] = E\left[(X - \mu)^2\right]$$

Squaring means the unit of variance is the **square** of the unit of $X$. If $X$ is in kg, then $V[X]$ is in kg².

---

### 3. Standard Deviation

$$\sigma[X] = \sqrt{V[X]} = \sigma$$

The standard deviation brings the unit **back to the same unit as $X$**. This is why standard deviation is often preferred in practice — it is directly interpretable on the same scale as your data.

**Geometrically**, $\sigma$ marks the distance from the mean to the **inflection points** of the bell curve — the points where the curve switches from concave to convex.

</div>

### Summary

| Property | Formula | Meaning |
|----------|---------|---------|
| Mean | E[X] = μ | Center / peak location |
| Variance | V[X] = σ² | Spread (squared units) |
| Std Dev | σ[X] = σ | Spread (original units) |

## Mean Squared Error (MSE)

### The Core Question

> **If I must summarize a random variable $X$ with one number $c$, which $c$ gives the smallest squared error on average?**

$$\boxed{c = E[X] \quad \longrightarrow \quad \textbf{The Mean}}$$

---

### Definition — Mean Squared Error (MSE) about $c$

For a random variable $X$ and $c \in \mathbb{R}$:

$$\text{MSE}(c) = E\left[(X - c)^2\right]$$

> 💡 We penalize the distance between $X$ and our chosen summary $c$, then average that squared penalty over all outcomes.

### Root Mean Squared Error (RMSE)

$$\text{RMSE}(c) = \sqrt{E\left[(X - c)^2\right]}$$

Much as standard deviation rescales variance to stay on the same scale as $X$, **RMSE rescales MSE** so that it stays on the same scale as $X - c$. Researchers often prefer to report RMSE rather than MSE.

---

### Connection to Variance & Raw Moments

| Special Case | Result | Name |
|--------------|--------|------|
| c = 0 | E[X²] | 2nd raw moment |
| c = E[X] | Var(X) | 2nd central moment |

---

### Theorem — Alternative Formula for MSE

For a random variable $X$ and $c \in \mathbb{R}$:

$$E\left[(X-c)^2\right] = \underbrace{V[X]}_{\text{fixed}} + \underbrace{(E[X] - c)^2}_{\text{minimized when } c = E[X]}$$

**Proof:**

$$E\left[(X-c)^2\right] = E\left[X^2 - 2cX + c^2\right] = E[X^2] - 2c\,E[X] + c^2$$

Adding and subtracting $E[X]^2$:

$$= \Big(E[X^2] - E[X]^2\Big) + \Big(E[X]^2 - 2c\,E[X] + c^2\Big)$$

$$= V[X] + \big(E[X] - c\big)^2 \quad \square$$

---

### Key Insight — The Bias-Variance Decomposition

| Term | Role | Controllable? |
|------|------|:---:|
| V[X] | Irreducible variance — noise in X | ❌ No |
| (E[X] − c)² | Squared bias — wrong guess error | ✅ Yes |

We can only control the **bias term**, and it equals zero exactly when $c = E[X]$.

> **Note:** A model can be bad because its prediction is very noisy (high variance), or because it is consistently wrong in one direction (high bias).
> - If a model always predicts 5 units too high → low variance, high bias
> - If predictions jump randomly around truth → high variance, low bias
> - MSE penalizes **both**.

### MSE Decomposition

$$\text{MSE} = \text{Var}(e) + (\text{Bias}(e))^2$$

$$\text{MSE}(c) = E[(X-c)^2] = \underbrace{V[X]}_{\text{unavoidable randomness}} + \underbrace{(E[X]-c)^2}_{\text{wrong guessing error}}$$

---

### Example

Suppose $E[X] = 10$, but you always guess $c = 7$.

$$(E[X] - c)^2 = (10 - 7)^2 = 9$$

This is the **"wrong guessing"** part — even if the data had zero randomness, you would still make an error because your chosen constant $c$ is not centered at the true mean.

> **The best constant guess is always the mean** — MSE is minimized when $c = E[X]$, and only the unavoidable variance remains.

---

### Theorem — The Expected Value Minimizes MSE

> For a random variable $X$, the value of $c$ that minimizes MSE is $c^* = E[X]$.

**Proof:**

$$\arg\min_{c\,\in\,\mathbb{R}}\; E\left[(X-c)^2\right] = \arg\min_{c\,\in\,\mathbb{R}} \left[V[X] + (E[X]-c)^2\right]$$

$$= \arg\min_{c\,\in\,\mathbb{R}} \;(E[X]-c)^2 \qquad (V[X] \text{ is fixed w.r.t. } c)$$

$$= E[X] \quad \square$$

> **In plain English:** If we had to pick one number as a prediction for $X$, the "best" choice (in terms of minimizing MSE) is $E[X]$.

---

### Summary

| Concept | Formula | Notes |
|---------|---------|-------|
| MSE | E[(X−c)²] | |
| RMSE | √E[(X−c)²] | Same scale as X−c |
| Alternative | V[X] + (E[X]−c)² | Bias-variance split |
| Optimal c | c* = E[X] | Minimizes MSE |
| Minimum MSE | V[X] | Irreducible floor |
| c = 0 case | E[X²] | 2nd raw moment |
| c = E[X] case | Var(X) | 2nd central moment |

> **Final MSE Intuition:** Under MSE, the best prediction is the number that balances all squared distances to the data. That balancing point is the mean.

## Covariance

Our goal is to describe the **relationship between random variables**. For example: if one quantity $X$ is high, is another $Y$ more or less likely to be high?

**Covariance** measures the extent to which two random variables *move together*.

$$\text{Cov}[X, Y] = E\big[(X - E[X])(Y - E[Y])\big]$$

| Sign of Cov[X,Y] | Meaning |
|------------------|---------|
| Positive | X and Y tend to move together |
| Negative | X large → Y tends to be small |
| Zero | No linear relationship |

> **Note:** Cov[·, ·] is an operator — so Cov[X, Y] is a **constant**, not a random variable.

---

### Theorem — Alternative Formula for Covariance

$$\boxed{\text{Cov}[X, Y] = E[XY] - E[X]\,E[Y]}$$

**Proof:**

$$\text{Cov}[X, Y] = E\big[(X - E[X])(Y - E[Y])\big]$$

$$= E\big[XY - X\,E[Y] - Y\,E[X] + E[X]E[Y]\big]$$

$$= E[XY] - E[X]E[Y] - E[X]E[Y] + E[X]E[Y]$$

$$= E[XY] - E[X]E[Y] \quad \square$$

---

### Theorem — Variance of Sums

Let $X$ and $Y$ be random variables. Then:

$$\boxed{V[X + Y] = V[X] + 2\,\text{Cov}[X, Y] + V[Y]}$$

More generally, $\forall\, a, b, c \in \mathbb{R}$:

$$\boxed{V[aX + bY + c] = a^2 V[X] + 2ab\,\text{Cov}[X, Y] + b^2 V[Y]}$$

**Proof:**

$$V[X + Y] = E[(X+Y)^2] - E[X+Y]^2$$

$$= E[X^2 + 2XY + Y^2] - (E[X] + E[Y])^2$$

$$= E[X^2] + 2E[XY] + E[Y^2] - E[X]^2 - 2E[X]E[Y] - E[Y]^2$$

$$= \big(E[X^2] - E[X]^2\big) + 2\big(E[XY] - E[X]E[Y]\big) + \big(E[Y^2] - E[Y]^2\big)$$

$$= V[X] + 2\,\text{Cov}[X, Y] + V[Y] \quad \square$$

---

### Theorem — Properties of Covariance

For random variables $X$, $Y$, $Z$, $W$ and constants $a, b, c, d \in \mathbb{R}$:

- **Constants have zero covariance:** $\text{Cov}[c, X] = \text{Cov}[X, c] = \text{Cov}[c, d] = 0$
- **Symmetry:** $\text{Cov}[X, Y] = \text{Cov}[Y, X]$
- **Self-covariance is variance:** $\text{Cov}[X, X] = V[X]$
- **Linearity:** $\text{Cov}[aX + c,\ bY + d] = ab\,\text{Cov}[X, Y]$
- **Bilinearity:** $\text{Cov}[X + W,\ Y + Z] = \text{Cov}[X,Y] + \text{Cov}[X,Z] + \text{Cov}[W,Y] + \text{Cov}[W,Z]$

**Proof:**

**Constants have zero covariance:**
$$\text{Cov}[c, X] = E[cX] - E[c]E[X] = cE[X] - cE[X] = 0$$
$$\text{Cov}[c, d] = E[cd] - E[c]E[d] = cd - cd = 0$$

**Symmetry:**
$$\text{Cov}[X, Y] = E[XY] - E[X]E[Y] = E[YX] - E[Y]E[X] = \text{Cov}[Y, X]$$

**Self-covariance is variance:**
$$\text{Cov}[X, X] = E[X^2] - E[X]^2 = V[X]$$

**Linearity:**
$$\text{Cov}[aX + c,\; bY + d] = E\big[a(X - E[X]) \cdot b(Y - E[Y])\big] = ab\,\text{Cov}[X, Y]$$

**Bilinearity:**
$$\text{Cov}[X+W,\; Y+Z] = \big(E[XY] - E[X]E[Y]\big) + \big(E[XZ] - E[X]E[Z]\big)$$
$$\quad + \big(E[WY] - E[W]E[Y]\big) + \big(E[WZ] - E[W]E[Z]\big)$$
$$= \text{Cov}[X,Y] + \text{Cov}[X,Z] + \text{Cov}[W,Y] + \text{Cov}[W,Z] \quad \square$$

> 💡 **Key insight:** $\text{Cov}[X, X] = V[X]$ tells us that **variance is a special case of covariance** — the covariance of a random variable with itself. Covariance is a natural generalization of variance.

---

### Covariance — Intuition with Real Data

$$\text{Cov}(X,Y) = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})$$

**Example — Experience vs. Salary:**

| Person | Experience X | Salary Y |
|--------|-------------|---------|
| A | 1 | 50 |
| B | 2 | 60 |
| C | 3 | 70 |
| D | 4 | 80 |

**Step 1 — Compute means:** $\bar{X} = 2.5$, $\bar{Y} = 65$

**Step 2 — Measure deviations:**

| Person | xi − X̄ | yi − Ȳ | Product |
|--------|--------|--------|---------|
| A | −1.5 | −15 | +22.5 |
| B | −0.5 | −5  | +2.5  |
| C | +0.5 | +5  | +2.5  |
| D | +1.5 | +15 | +22.5 |

$$\text{Cov}(X,Y) = \frac{22.5 + 2.5 + 2.5 + 22.5}{4} = +12.5$$

Every row contributes a positive value because low experience pairs with low salary, and high experience pairs with high salary.

---

### The Real Intuition

> "When one variable moves away from its mean, does the other move in the **same direction**?"

| Case | Signs | Product | Effect on Cov |
|------|-------|---------|---------------|
| Both above mean | (+)(+) | positive | Cov ↑ |
| Both below mean | (−)(−) | positive | Cov ↑ |
| Opposite sides | (+)(−) | negative | Cov ↓ |

| Sign | Meaning | Example |
|------|---------|---------|
| Positive | Move together | Experience ↑ → Salary ↑ |
| Negative | Move opposite | Price ↑ → Demand ↓ |
| Near-zero | No pattern | Products cancel |

> If data lies perfectly on a line, every product has the same sign — they all reinforce rather than cancel — producing the largest possible covariance.

## Correlation

> **Correlation** is a rescaled version of covariance that captures linear dependence between two random variables, normalized to always fall between −1 and +1.

### Definition — Correlation

For random variables $X$ and $Y$ with $\sigma[X] > 0$ and $\sigma[Y] > 0$:

$$\boxed{\rho[X, Y] = \frac{\text{Cov}[X, Y]}{\sigma[X]\,\sigma[Y]}}$$

| ρ | Meaning |
|---|---------|
| +1 | Perfect positive linear relationship |
| −1 | Perfect negative linear relationship |
| 0 | No linear correlation |

Correlation is bounded: $-1 \leq \rho[X, Y] \leq 1$ — making it easier to interpret than raw covariance which has no fixed scale.

---

### From Covariance to Correlation

Covariance depends on the **scale and units** of the variables — pairing salary in dollars with height in centimeters produces a huge covariance simply because the numbers are large, not because the relationship is strong. Correlation fixes this by dividing by the typical spread of both variables.

---

### ⚠️ Important Note About Correlation

Correlation does **not** tell us how much one variable numerically changes another.

A correlation of $0.8$ does **not** mean a 1-unit increase in $X$ causes an 80% increase in $Y$. Instead, correlation measures how strongly and consistently two variables **move together in a linear way**.

A high positive correlation means:
- when $X$ is above its mean, $Y$ also tends to be above its mean
- when $X$ is below its mean, $Y$ also tends to be below its mean

> Correlation measures **synchronization of movement**, not the amount of change.
> The quantity that measures "how much $Y$ changes when $X$ changes by 1 unit" is the **slope in regression**, not correlation.

---

### The Tangible Intuition

| Question | Answered by |
|----------|-------------|
| Do they move together? | Covariance |
| How much does each fluctuate on its own? | Std deviation |
| How strong is co-movement vs. own variability? | Correlation |

$$\rho(X,Y) = \frac{\text{co-movement}}{\text{spread of } X \times \text{spread of } Y}$$
