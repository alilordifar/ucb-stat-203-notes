## 1. Continuous Random Variables

A continuous random variable arises when measurements can take essentially any precision (height, weight, time) — $X(\Omega)$ is an uncountable set (an interval or union of intervals of $\mathbb{R}$). Loosely: a random variable is continuous if its CDF is a continuous function.

> **Definition — Continuous Random Variable**
> A random variable $X$ is *continuous* if there exists a non-negative function $f: \mathbb{R} \to \mathbb{R}$ such that the CDF of $X$ is
>
> $$F(x) = \Pr[X \le x] = \int_{-\infty}^{x} f(u)\,du, \quad \forall x \in \mathbb{R}$$

$f$ is called the **probability density function (PDF)**. By the Fundamental Theorem of Calculus, the PDF is unique.

## 2. Probability Density Function (PDF)

> 💡 **PDF:** Tells you how dense probability is around a value — how concentrated values are. It is **NOT** the probability itself.

> **Definition — Probability Density Function (PDF)**
> For a continuous random variable $X$ with CDF $F$, the *probability density function* of $X$ is
>
> $$f(x) = \left.\frac{dF(u)}{du}\right|_{u=x}, \quad \forall x \in \mathbb{R}$$

A PMF gives the "jump" size at a point; a PDF gives the instantaneous slope of the CDF at a point. For small $\epsilon > 0$, moving from $x$ to $x+\epsilon$ changes the CDF by approximately $\epsilon f(x)$.

---

## 3. Properties of PDFs

> **Theorem — Properties of PDFs**
> For a continuous random variable $X$ with PDF $f$:
> - $\forall x \in \mathbb{R}, f(x) \geq 0$ → *PDF can never be negative*
> - $\displaystyle\int_{-\infty}^{\infty} f(x)\,dx = 1$

**Proof:** $F$ is nondecreasing, so $f(x) = \left.\dfrac{dF(u)}{du}\right|_{u=x} \geq 0$, and

$$\int_{-\infty}^{\infty} f(x)\,dx = \lim_{x\to\infty}\int_{-\infty}^{x} f(u)\,du = \lim_{x\to\infty} F(x) = 1 \qquad \blacksquare$$

## 4. Event Probabilities

> **Theorem — Event Probabilities for Continuous Random Variables**
> For a continuous random variable $X$ with PDF $f$:
> - $\forall x \in \mathbb{R}, \Pr[X=x] = 0$ → *prob of a single point*
> - $\forall x \in \mathbb{R}, \Pr[X<x] = \Pr[X\le x] = F(x) = \displaystyle\int_{-\infty}^{x} f(u)\,du$
> - $\forall x \in \mathbb{R}, \Pr[X>x] = \Pr[X\ge x] = 1-F(x) = \displaystyle\int_{x}^{\infty} f(u)\,du$
> - $\forall a,b \in \mathbb{R}$ with $a \le b$:
>   $$\Pr[a<X<b] = \Pr[a\le X<b] = \Pr[a<X\le b] = \Pr[a\le X\le b] = F(b)-F(a) = \int_a^b f(x)\,dx$$

**Proof:**

*Step 1 — $\Pr[X=x]=0$ (by contradiction):* Suppose $\Pr[X=x] = P(\{X=x\}) > 0$. Since $P$ is continuous and $\lim_{u\to x^-}\{X\le u\} = \{X<x\}$:

$$\lim_{u\to x^-} F(u) = P(\{X<x\}) < P(\{X<x\}) + P(\{X=x\}) = P(\{X\le x\}) = F(x)$$

This means $F$ has a discontinuity at $x$ — contradicting continuity of $X$. So $\Pr[X=x] = 0$.

*Step 2 — CDF forms:* Since $\Pr[X=x]=0$:

$$\Pr[X<x] = P(\{X<x\}) = P(\{X\le x\}) = F(x) = \int_{-\infty}^x f(u)\,du$$

Similarly:

$$\Pr[X\ge x] = P(\{X>x\}) + P(\{X=x\}) = \Pr[X>x] = 1-F(x) = \int_{-\infty}^{\infty} f(u)\,du - \int_{-\infty}^x f(u)\,du = \int_x^{\infty} f(u)\,du$$

*Step 3 — interval probabilities:* Since $\{X<a\}\cup\{a\le X\le b\} = \{X\le b\}$ and the two are disjoint, countable additivity gives:

$$P(\{a\le X\le b\}) = P(\{X\le b\}) - P(\{X<a\})$$

$$\Pr[a\le X\le b] = F(b)-F(a) = \int_a^b f(x)\,dx \qquad \blacksquare$$

*(final equality follows from the Fundamental Theorem of Calculus)*

> **Note:** By countable additivity, this extends to compute $\Pr[X \in D]$ for any $D$ that is a countable union of disjoint intervals in $\mathbb{R}$.

## 5. Uniform Distribution

All possible outcomes are equally likely — no outcome or region is preferred.

### Standard Uniform — $U(0,1)$

$$
f(x) =
\begin{cases}
1 & : 0 \le x \le 1 \\
0 & : \text{otherwise}
\end{cases}
\qquad
F(x) =
\begin{cases}
0 & : x<0 \\
x & : 0\le x\le 1 \\
1 & : x>1
\end{cases}
$$

- PDF is constant → area under it grows **linearly** with width
- CDF is the area under the PDF from the start up to $x$ → probability accumulates at a constant rate

### General Uniform — $[A,B]$

$$
f(x) =
\begin{cases}
\dfrac{1}{B-A} & : A \le x \le B \\
0 & : \text{otherwise}
\end{cases}
$$

**Derivation of the CDF**, for $A \le x \le B$:

$$F(x) = \int_{-\infty}^{x} f(y)\,dy = \int_A^x \frac{1}{B-A}\,dy = \left.\frac{y}{B-A}\right|_A^x = \frac{x-A}{B-A}$$

$$
F(x) =
\begin{cases}
0 & : x < A \\
\dfrac{x-A}{B-A} & : A \le x < B \\
1 & : x \ge B
\end{cases}
$$

### Examples

**Discrete Uniform** — fair die: outcomes $1,2,3,4,5,6$, each $p = \tfrac{1}{6}$

**Continuous Uniform** — a 10 m hallway; drop a coin randomly along it. Any 1 m section is equally likely to get the coin. No exact point (e.g. 3.27 m) has probability by itself — only intervals matter.

$$\text{density} = \frac{1}{10}, \qquad \Pr(2 < X < 3) = \int_2^3 \frac{1}{10}\,dx = \frac{1}{10}$$

> 💡 **Point:** In a continuous uniform distribution, probability is spread evenly over space.

## 6. Normal Distribution

A continuous R.V. $X$ has a normal distribution with parameters $\mu$ and $\sigma$ (or $\mu$ and $\sigma^2$), where $-\infty < \mu < \infty$ and $\sigma > 0$, if the PDF is:

$$f(x;\mu,\sigma) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-(x-\mu)^2/(2\sigma^2)}, \quad -\infty < x < \infty$$

- $E(X) = \mu$, $V(X) = \sigma^2$ → parameters are the mean and standard deviation of $X$
- Abbreviated: $X \sim N(\mu, \sigma^2)$

> **Definition:** A normal distribution is a continuous probability distribution where values cluster around a mean, the distribution is symmetric, and values farther from the mean become exponentially less likely.

### Parameters
- $\mu$ is both the **mean** and the **median** — changing it moves the curve left or right
- $\sigma$ is the **scale parameter** — it stretches the curve horizontally

### CDF Behavior (recall Unit 2, p.55)
1. The CDF shows accumulated probability
2. The slope of the CDF at each point tells you how fast probability is accumulating there
3. Where the CDF is steepest (around the mean), the derivative is largest → peak of the bell curve
4. Where the CDF flattens (tails), the derivative is small → thin tails of the bell

### Empirical Rule (68–95–99.7)
- 68% of the area is within 1 standard deviation of the mean
- 95% is within 2 standard deviations
- 99.7% is within 3 standard deviations

### Rules of Thumb (helpful to memorize)
1. Useful when a variable is approximately normally distributed
2. Expect ~68% of values within 1 standard deviation
3. Very few values are more than 2 standard deviations from the mean (important for hypothesis testing)

## 7. Support

The set of values a random variable can actually take with non-zero probability.

> **Definition — Support**
> For a random variable $X$ with PMF/PDF $f$, the *support* of $X$ is
>
> $$\text{Supp}[X] = \{x \in \mathbb{R} : f(x) > 0\}$$

## 🧾 Summary Cheat Sheet

**PDF** — density around a value, not probability itself
$$f(x) = \frac{dF(x)}{dx}$$

**CDF** — accumulated probability up to $x$
$$F(x) = \Pr[X \le x]$$

**Standard Uniform** — constant density on $[0,1]$

**General Uniform** — constant density $\tfrac{1}{B-A}$ on $[A,B]$

**Normal** — bell curve, $\mu$ shifts location, $\sigma$ stretches spread
$$X \sim N(\mu, \sigma^2)$$

**Support** — where $f(x) > 0$
