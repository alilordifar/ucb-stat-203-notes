## 1. Discrete Random Variables

A discrete random variable takes on a **finite** or **countably infinite** number of different values.

> **Definition — Discrete Random Variable**
> A random variable $X$ is *discrete* if its range, $X(\Omega)$, is a countable set.

---

## 2. Probability Mass Function (PMF)

Given a discrete random variable $X$, we summarize the probability of each outcome $x$ with a **probability mass function (PMF)**.

> **Definition 1.2.5. Probability Mass Function (PMF)**
> For a discrete random variable $X$, the *probability mass function* of $X$ is
>
> $$f(x) = \Pr[X=x], \quad \forall x \in \mathbb{R}$$

> 💡 **Point:** For a discrete $X$, the PMF tells us everything about its distribution — loosely, the collection of probabilities assigned to events defined in terms of $X$. **PMF = probability at one value.**

## 3. Data Distribution vs. Probability Distribution

| | Probability Distribution | Data Distribution |
|---|---|---|
| Nature | Theoretical | Observed |
| Describes | The population | The sample |
| Type | Mathematical model | Actual collected data |
| Use | Compute probabilities | Estimate the probability distribution |

**Analogy:** Probability distribution = the recipe. Data distribution = the cake you baked. You only see the cake — from it, you try to infer the recipe.

```
Probability Distribution
        │
        ▼
   Generate Samples
        │
        ▼
   Observed Dataset
        │
        ▼
   Data Distribution
```

**Example:** Heights of 5 people — `170, 175, 180, 180, 185` is the **data distribution** (what you observed). If millions of people's heights follow a bell curve, that underlying Normal curve is the **probability distribution**. More data → data distribution converges toward the true probability distribution.

## 4. The Distribution

A distribution describes how probability is spread across all possible values.

- A **PMF** gives the probability at each single point.
- The collection of all those point probabilities together is the **probability distribution**.

| Individual probabilities | The collection of them |
|---|---|
| = values of the PMF | = the distribution |

## 5. Named Discrete Distributions

### Bernoulli
Models a single experiment with only two outcomes (success/failure), success probability $p$.

$$
f(x) =
\begin{cases}
1-p & : x=0 \\
p & : x=1 \\
0 & : \text{otherwise}
\end{cases}
$$

**Example:** Biased coin — $X=1$ (tails) w.p. $p$, $X=0$ (heads) w.p. $1-p$.

### Binomial
Number of successes in a fixed number of independent yes/no trials, same success probability each trial.

$$P(X=x) = \binom{n}{x} p^x (1-p)^{n-x}, \quad x=0,1,\dots,n \qquad \text{where } \binom{n}{x} = \frac{n!}{x!(n-x)!}$$

### Geometric
Number of trials until the **first** success.

$$
f(x) =
\begin{cases}
(1-p)^{x-1}p & : x \in \mathbb{N} \\
0 & : \text{otherwise}
\end{cases}
$$

**Example:** Fair coin ($p=\tfrac12$) — first tails on flip 1 w.p. $\tfrac12$, flip 2 w.p. $\tfrac14$, flip 3 w.p. $\tfrac18$, ...

### Binomial vs. Geometric — quick contrast

| | Question it answers |
|---|---|
| **Binomial** | How many successes occur in a fixed number of trials? |
| **Geometric** | How many trials are needed until the first success? |

## 6. Properties of PMFs

> **Theorem — Properties of PMFs**
> For a discrete random variable $X$ with PMF $f$:
> - $\forall x \in \mathbb{R},\; f(x) \geq 0$
> - $\displaystyle\sum_{x \in X(\Omega)} f(x) = 1$

## 7. Cumulative Distribution Function (CDF)

> **Definition — Cumulative Distribution Function (CDF)**
> For a random variable $X$, the *cumulative distribution function* of $X$ is
>
> $$F(x) = \Pr[X \le x], \quad \forall x \in \mathbb{R}$$

The CDF returns the probability that an outcome will be **less than or equal to** a given value — i.e., "probability up to a value."

---

## 8. Properties of CDFs

> **Theorem — Properties of CDFs**
> For a random variable $X$ with CDF $F$:
> - $F$ is nondecreasing: $\forall x_1,x_2 \in \mathbb{R}$, if $x_1<x_2$ then $F(x_1)\le F(x_2)$
> - $\lim_{x\to-\infty} F(x) = 0$
> - $\lim_{x\to\infty} F(x) = 1$
> - $\forall x\in\mathbb{R},\; 1-F(x) = \Pr[X>x]$

**Proof:**

Let $x_1 < x_2$. Since $\{X\le x_1\} \subseteq \{X\le x_2\}$, monotonicity gives:

$$F(x_1) = \Pr[X\le x_1] \le \Pr[X\le x_2] = F(x_2)$$

Since $\lim_{x\to-\infty}\{X\le x\} = \emptyset$:

$$\lim_{x\to-\infty} F(x) = P(\emptyset) = 0$$

Since $\lim_{x\to\infty}\{X\le x\} = \Omega$:

$$\lim_{x\to\infty} F(x) = P(\Omega) = 1$$

By the complement rule:

$$1-F(x) = 1-P(\{X\le x\}) = P(\{X\le x\}^C) = P(\{X>x\}) = \Pr[X>x] \qquad \blacksquare$$

## 🧾 Summary Cheat Sheet

**PMF** — probability at one exact value
$$f(x) = \Pr[X=x]$$

**CDF** — probability up to a value
$$F(x) = \Pr[X\le x]$$

**Bernoulli** — success/failure, single trial, parameter $p$

**Binomial** — # successes in $n$ trials
$$\binom{n}{x}p^x(1-p)^{n-x}$$

**Geometric** — # trials until 1st success
$$(1-p)^{x-1}p$$
