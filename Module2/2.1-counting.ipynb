## 1. Counting Basics

**Counting:** If one thing can be done in $n$ ways and another thing can be done in $m$ ways, then doing them together can be done in $m \times n$ ways.

**Multiplication Rule:** If we have an experiment with $n_1$ possible outcomes, and for each outcome of the 1st experiment there are $n_2$ outcomes for the 2nd experiment, ..., and for each there are $n_r$ outcomes for the $r$th experiment, then overall there are

$$n_1 \cdot n_2 \cdots n_r$$

possible outcomes for the combined experiment (doing all these smaller experiments together).

**Example — Ice cream:**
```
Ice Cream ──► Cone Type: C (Cake), W (Waffle)
                   │
                   ▼
              Flavor: Chocolate, Vanilla, Strawberry
```
- $6 = 3 \times 2$ (flavor first)
- $6 = 2 \times 3$ (cone first) → you can choose flavor first, order doesn't matter

**Ex:** Find the probability of a full house in poker (5-card hand).
> Full house = three cards of one rank + two of another (e.g., 3 sevens and 2 tens).

## 2. Binomial Coefficient

$$\binom{n}{k} = \frac{n!}{(n-k)!\,k!}; \quad = 0 \text{ if } k > n$$

> Directly comes from the multiplication rule. (If you have 10 people, you can't choose 11 from them.)

**What it represents:** the number of ways to choose a subset of size $k$ out of $n$ people, where **order doesn't matter**.

**Derivation:**
$$\binom{n}{k} = \frac{n(n-1)(n-2)\cdots(n-k+1)}{k!} = \frac{n!}{(n-k)!\,k!}$$

**Full house example:**
- Need 3 out of 4 sevens, 2 out of 4 tens:
$$\frac{13\binom{4}{3} \cdot 12\binom{4}{2}}{\binom{52}{5}}$$
- $13$ = choices for the triple's rank, $12$ = remaining choices for the pair's rank (since the rank is already picked)
- $\binom{52}{5}$ = all possible 5-card outcomes

## 3. Sampling Table — Choose $k$ objects out of $n$

| | Order matters | Order doesn't matter |
|---|:---:|:---:|
| **With replacement** | $n^k$ | $\dbinom{n+k-1}{k}$ |
| **Without replacement** | $\dfrac{n!}{(n-k)!}$ | $\dbinom{n}{k} = \dfrac{n!}{k!(n-k)!}$ |

*(Without replacement, order matters = $n(n-1)\cdots(n-k+1)$)*

### Permutations
Given $n$ distinct objects, arrange $r$ of them in a straight line:

$$_nP_r = n(n-1)(n-2)\cdots(n-r+1)$$

**Examples:**
- $_4P_2 = 4 \times 3$
- $_7P_3 = 7 \times 6 \times 5$

---

## 4. Permutation Formula Derivation & Multinomial Coefficients

$$_nP_r = n(n-1)(n-2)\cdots(n-(r-1))$$

$$n! = \underbrace{n(n-1)(n-2)\cdots(n-(r-1))}_{_nP_r} \cdot \underbrace{(n-(r-1)-1)!}_{(n-r)!}$$

$$n! = {_nP_r} \cdot (n-r)! \;\;\Longrightarrow\;\; \boxed{_nP_r = \dfrac{n!}{(n-r)!}}$$

### Permutations with Identical Objects

Suppose a set of $n$ objects has $n_1$ of one type (indistinguishable from each other), $n_2$ of a second type, ..., $n_k$ of a $k$-th type, where

$$n = n_1 + n_2 + \cdots + n_k$$

**Step 1:** If all $n$ objects were different, the number of permutations is simply

$$n! = n(n-1)(n-2)\cdots 1$$

**Step 2 — Problem with identical objects:** Now suppose all objects of the same type are indistinguishable ($n_1$ of type 1, $n_2$ of type 2, ..., $n_k$ of type $k$).

## 5. Correcting for Overcounting (Multinomial Coefficient)

> **Note:** When we use $n!$, we are overcounting — swapping identical objects doesn't create a new arrangement.

**Step 3 — How much are we overcounting?**
For the $n_1$ identical objects, among themselves they can be arranged in $n_1!$ ways — but all look the same. So every distinct arrangement is counted $n_1! \cdot n_2! \cdot n_3! \cdots n_k!$ times.

**Step 4 — Correct by dividing:**

$$\boxed{\;_n P_{n_1, n_2, \dots, n_k} = \dfrac{n!}{n_1!\, n_2! \cdots n_k!}\;} \quad \text{(distinct permutations)}$$

**Example — AAB** ($n=3$, $n_1=2$ A's, $n_2=1$ B):
```
A₁A₂B, A₂A₁B  → same           (2 arrangements collapse to 1)
A₂BA₁, A₁BA₂  → same           (2 arrangements collapse to 1)
BA₁A₂, BA₂A₁  → same           (2 arrangements collapse to 1)
```
→ 3 distinct arrangements: $\dfrac{3!}{2!\,1!} = 3$

**Example — ABC** (all distinct, $n_1=n_2=n_3=1$):
```
ABC, ACB, BAC, BCA, CAB, CBA → 6 ways
```
Each group of $k$ identical items has $k!$ internal orderings — here all groups have size 1, so $\dfrac{6}{1\cdot1\cdot1} = 6$ (no overcounting since nothing repeats).

## 6. Combinations (Order Does Not Matter)

**Example:** $n = \{A, B, C, D\}$, $k = 2$

```
A first: AB, AC, AD
B first: BA, BC, BD
C first: CA, CB, CD
D first: DA, DB, DC
```
Total (permutations): $4 \times 3 = 12$

Now switch to **combinations** — technically, combinations remove duplicates caused by order. $AB$ and $BA$ are the same group, so we collapse each "mirror" pair into one:

$$\frac{12}{2!} = 6 \quad \text{(overcounted by a factor of } k! = 2!\text{)}$$

## 7. Practical Tips

1. **Don't lose common sense.** $\binom{52}{5}$ is self-annotating.
2. **Check answers**, especially by trying **simple and extreme cases**.
3. **Label** people, objects, etc. — if you have $n$ people, label them $1, 2, \dots, n$.

**Example — Split 10 people into a team of 6 and a team of 4:**

$$\binom{10}{4} = \binom{10}{6} \;\Rightarrow\; \frac{10!}{4!\,6!} = \frac{10!}{6!\,4!}$$

- $\binom{10}{4}$: choose the team of 4, whoever's left is the team of 6.
- $\binom{10}{6}$: choose the team of 6, whoever's left is the team of 4.

**Example — 2 teams of 5** (teams are *unlabeled*, so divide by 2 to avoid double-counting which team is "team A" vs "team B"):

$$\frac{\binom{10}{5}}{2}$$

## 8. Stars and Bars — With Replacement, Order Doesn't Matter

**Setup:** Pick $k$ times from a set of $n$ objects, where order does not matter, **with replacement**:

$$\binom{n+k-1}{k} \text{ ways}$$

### Extreme cases (sanity checks)
- $k = 0 \Rightarrow \binom{n-1}{0} = 1$
- **Why is $0! = 1$?** Think of it as: if you have $n$ people lined up for ice cream, how many ways can you order them? If no one is there, there's exactly **one** way — the way where no one is there.
- $k = 1 \Rightarrow \binom{n}{1} = n$ (a less extreme case)
- $n = 2$ (simplest non-trivial example), picking $k$ times:
$$\binom{k+1}{k} = \binom{k+1}{1} = k+1$$
(Objects: e.g., 7 dots split between obj1 and obj2.)

---

## 9. Stars and Bars — Equivalent Formulation & Story Proof

**Equivalent problem:** How many ways are there to put $k$ indistinguishable particles into $n$ distinguishable boxes?

```
Example: n = 4, k = 6
[○○○] | [   ] | [○○] | [○]
```

- $k$ dots (stars), $n-1$ separators (bars)

$$\binom{n+k-1}{k} = \binom{n+k-1}{n-1}$$

> Choose where the dots are, and the separators' positions are automatically determined.

*(Historical note: Bose-Einstein condensate, 1920s — same combinatorial structure.)*

### Story Proof
A **story proof** is a proof by interpretation rather than by algebra or calculus.

**Example (symmetry):**
$$\binom{n}{k} = \binom{n}{n-k}$$

**Example 2 (president selection):**
$$n\binom{n-1}{k-1} = k\binom{n}{k}$$

- **Right side:** pick $k$ people out of $n$, with one designated as president.
- **Left side:** first block off the president ($n$ ways), then choose the remaining $k-1$ committee members from the other $n-1$ people.
- Both count the same thing in two different ways.

## 10. Stars and Bars — Worked Example (Ice Cream Scoops)

**Setup:** Order does not matter, with replacement.

**Step 1:** Picking 5 scoops of ice cream from 3 flavors.
$$n = 3 \;(\text{strawberry, chocolate, vanilla}), \quad k = 5$$

**Step 2:** Let $x_1$ = number of times we pick item 1 (strawberry), $x_2$ = number of times we pick item 2, ..., $x_n$ = number of times we pick item $n$.

$$k = x_1 + x_2 + \cdots + x_n, \quad \text{where each } x_i \geq 0$$

**Step 3:** Convert to stars and bars (dividers):
- Each pick = a star (★)
- To separate into $n$ groups, place $n-1$ bars (|)
- Total symbols = $k + (n-1) = n+k-1$
- Any solution corresponds to arranging $k$ stars and $n-1$ bars:

$$\binom{n+k-1}{k} = \binom{7}{5} = \frac{7!}{2!\,5!} = 21$$

**Step 4:** Count arrangements — total symbols $= k + (n-1) = n+k-1$.

## 11. Vandermonde's Identity

**Note:** We can pick who's in the club ($n$ choose $k$), then one of those $k$ must be elected president → multiply by $k$ (multiplication rule).

**OR:** First choose the president, then choose $k-1$ more members from the remaining $n-1$ people.

**Vandermonde's Identity:**

$$\binom{m+n}{k} = \sum_{j=0}^{k} \binom{m}{j}\binom{n}{k-j}$$

```
m people: pick j
n people: pick k-j
```

**Example:** Want $k=5$, so split as $j=2$ from $m$-group and $k-j=3$ from $n$-group.

$$\binom{9}{3} = \frac{9!}{3!\,6!} = \frac{9 \times 8 \times 7}{3 \times 2 \times 1} = 84$$

**Enumeration example** ($n=4$ types, $k=6$ total items) — all ways to distribute 6 indistinguishable items among 4 distinguishable categories $(x_1,x_2,x_3,x_4)$ with $\sum x_i = 6$:

```
(6,0,0,0)  (2,4,0,0)  (0,3,3,0)  (0,0,4,2)
(5,1,0,0)  (1,5,0,0)  (0,2,4,0)  (0,0,3,3)
(4,2,0,0)  (0,6,0,0)  (0,1,5,0)  (0,0,2,4)
(3,3,0,0)  (0,5,1,0)  (0,0,6,0)  (0,0,1,5)
(0,4,2,0)  (0,0,5,1)  (0,0,0,6)
```
