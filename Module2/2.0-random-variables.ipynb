## 1. Random Variables

**Core idea:** A random variable maps outcomes of a random process to real numbers, so we can analyze uncertainty using math.

> Many aspects of the world can be quantified. When a random generative process has outcomes that can be assigned real number values, its properties can more easily be described mathematically. A random variable is a variable that takes on a real value that is determined by a random generative process — it lets us represent random events by mapping those events to plain numbers.

We use random variables to turn uncertain outcomes into numbers so we can analyze, calculate probabilities, and make predictions using math.

### Why it matters
| # | Benefit |
|---|---------|
| 1 | Apply probability rules and formulas |
| 2 | Compute expectation, variance, and risk |
| 3 | Connect real-world randomness to mathematical models |

> 💡 **Key point:** Without random variables, probability stays descriptive; with them, it becomes quantitative and usable.

### Formal Definition

A random variable is a **function** mapping each outcome in the sample space to a number:

$$X: \Omega \rightarrow \mathbb{R}$$

| Term | Meaning |
|---|---|
| **Independent variable** | The outcome of the experiment (an element of sample space $\omega \in \Omega$) |
| **Dependent variable** | The value of the random variable, $X(\omega)$ |

**Tangible example — coin flip:**
```
Outcome = Heads        → input   (ω)
X(Heads) = 1            → output  (X(ω))
```

> ⚠️ **Subtlety:** Formally, a random variable is a real-valued *function* of the outcome of a random process. Technically it's neither random nor a variable — it's just a function.

## 2. What is an Operator?

**Intuition:** Think of an operator as a **machine** that takes a random variable as input and outputs **one single real number**.

```
Random Variable X
        │
        ▼
+------------------+
|     Operator     |
+------------------+
        │
        ▼
  One Real Number
```

### Example — Random Variable (die roll)

| Outcome (ω) | X(ω) |
|---|---|
| Die shows 1 | 1 |
| Die shows 2 | 2 |
| Die shows 3 | 3 |
| Die shows 4 | 4 |
| Die shows 5 | 5 |
| Die shows 6 | 6 |

### Example 1 — Expectation (Mean)

$$E[X] = 3.5$$

```
{1,2,3,4,5,6} ──► Expectation ──► 3.5
```

### Example 2 — Variance

$$Var(X) = 2.9167$$

```
{1,2,3,4,5,6} ──► Variance ──► 2.9167
```

### Example 3 — Maximum

$$X = \{4, 7, 2, 9\} \quad\Rightarrow\quad A[X] = \max(X) = 9$$

### Why "summarize"?

A random variable carries **a lot of information** (e.g., all 6 die values). An operator extracts **one property**:

| Operator | What it tells you |
|---|---|
| Expectation | "On average, the value is 3.5." |
| Variance | "The spread of the values is 2.9167." |
| Maximum | "The largest possible value is 6." |

> **Textbook definition, plainly:** An operator $A$ takes a random variable $X(\cdot)$ and maps it to a real number.

### Programming Analogy (Python)

```python
numbers = [1, 2, 3, 4, 5, 6]

len(numbers)   # 6
sum(numbers)   # 21
max(numbers)   # 6
min(numbers)   # 1
```

Same idea as:
```
E[X]
Var(X)
P(X > 3)
```
— each takes the **entire random variable** and returns **one number**.

### 🎯 One-Sentence Definition
> **An operator is a mathematical function that takes an entire random variable as input and returns one real number summarizing some property of it.**

Common operators: Expectation · Variance · Standard Deviation · Probability · Min · Max · Median

## 3. Random Variable vs. Random Vector vs. Bivariate Relationship

### Random Variable — one measurement
```
Person ──► Height = X
```
Each person has **one value** for this measurement.

### Random Vector — multiple random variables grouped together

$$
\mathbf{X}=
\begin{bmatrix}
\text{Height}\\
\text{Weight}\\
\text{Age}
\end{bmatrix}
$$

| Person | Height | Weight | Age |
|---|---:|---:|---:|
| A | 180 | 75 | 30 |
| B | 165 | 60 | 25 |

> Each **row** = one random vector (one observation, multiple features).

### Bivariate Relationship — relationship between two random variables

```
Height ─────────► Weight
      (Are they related?)
```

Example questions: Do taller people weigh more? Does study time affect exam scores? Does temperature affect electricity usage?

| Height | Weight |
|---:|---:|
| 170 | 65 |
| 175 | 70 |
| 180 | 78 |
| 185 | 84 |

→ As height increases, weight tends to increase = **positive bivariate relationship**.

## 🧾 Summary Cheat Sheet

| Concept | Definition | Data Science Analogy |
|---|---|---|
| **Random Variable** | One measurement (e.g., Height) | One feature (one column) |
| **Random Vector** | Multiple measurements grouped (e.g., Height, Weight, Age) | One data record (all columns, one row) |
| **Bivariate Relationship** | Relationship between two random variables | Correlation / covariance / regression between two columns |
| **Operator** | Function mapping a random variable → one real number | Summary statistic (`mean()`, `var()`, `max()`) |
