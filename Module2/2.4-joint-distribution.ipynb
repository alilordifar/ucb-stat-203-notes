## 1. Equality of Random Variables

Two random variables are identical if and only if they produce the **same value for every outcome** $\omega$ in the sample space.

> **Definition — Equality of Random Variables**
> Let $X$ and $Y$ be random variables. Then $X=Y$ if, for all $\omega \in \Omega$, $X(\omega)=Y(\omega)$.

**Don't assume $X=Y$ just because:**

1. **$X$ and $Y$ have the same distribution**
   - $X=1$ if Paul's coin lands heads, 0 otherwise
   - $Y=1$ if Alex's coin lands heads, 0 otherwise
   - $f_X = f_Y$ but $X \neq Y$

2. **$P(X=Y)=1$**
   - $X=1$ if Paul climbs Half Dome, 0 otherwise
   - $Y=1$ if Alex climbs El Capitan, 0 otherwise
   - $P(X=Y)=1$ but $X \neq Y$ → they are different events

## 2. Joint Distribution — Intuition

Tells you the probability of multiple R.V.s happening together — the chance $X$ is this **and** $Y$ is that, at the same time.

**Conditional probability:** $P(A \mid B)$ — probability $A$ happens, given $B$ already happened.

**Marginal vs. Joint:**
- **Marginal** — "how often does $X$ happen?" / "how often does $Y$ happen?" — each looked at by itself.
- **Joint** — "when $X$ happens, what tends to happen to $Y$?" — captures the connection between them.

> 💡 Knowing only the marginals is usually not enough to know the connection — many different joint patterns can produce the same individual (marginal) frequencies.

**Independence** is the special case with **no connection** — knowing $X$ tells you nothing about $Y$; "together" is just the product of the individuals.

> **Summary:** Marginals describe individual behavior, the joint describes interaction. Only under independence do marginals tell the whole story.

## 3. Marginal vs. Joint — Recoverability

A **marginal distribution** describes each variable individually; a **joint distribution** describes how they occur *together*. We cannot recover the joint from the marginals because the marginals lose the pairing information.

**Example:** Marginals say 50% of people are 170 cm, 50% are 180 cm; separately, 50% weigh 60 kg, 50% weigh 80 kg. These marginals don't tell us whether 170 cm people weigh 60 kg or 80 kg. Both pairings below have the *same marginals*:
- (170,60) & (180,80)
- (170,80) & (180,60)

**Setup:** $\Omega$ maps to $B$ (buy / not buy) and $D$ (test drive).

### From Joint → Marginal

$$f_B(b) = P(B=b) = \sum_{d \in \text{Supp}[D]} f(b,d) \quad \text{for all } b \in \mathbb{R}$$

```
Joint distribution table

              B=0    B=1   | f_D
    D=0       0.1    0.1   | 0.2
    D=1       0.4    0.1   | 0.5
    D=2       0.1    0.2   | 0.3
    --------------------------
    f_B       0.6    0.4
```

We just **marginalized** by summing across rows/columns.

> Just because we measured another variable doesn't mean we can no longer map down to less information.

### From Marginal → Joint? — **No.**

Two different joints can share the same marginals:

```
Joint A (original)

              B=0    B=1   | f_D
    D=0       0.1    0.1   | 0.2
    D=1       0.4    0.1   | 0.5
    D=2       0.1    0.2   | 0.3
    --------------------------
    f_B       0.6    0.4
```

```
Joint B (different joint, same marginals)

              B=0    B=1   | f_D
    D=0       0.2    0.0   | 0.2
    D=1       0.2    0.3   | 0.5
    D=2       0.2    0.1   | 0.3
    --------------------------
    f_B       0.6    0.4
```

> 💡 **Point:** The marginals alone don't determine the joint — the joint captures the interaction that the marginals lose.

## 4. Discrete Bivariate Distributions

> **Definition — Joint PMF**
> $$f(x,y) = \Pr[X=x, Y=y], \quad \forall x,y \in \mathbb{R}$$

> **Definition — Joint CDF**
> $$F(x,y) = \Pr[X\le x, Y\le y], \quad \forall x,y \in \mathbb{R}$$

The marginal distribution of a variable tells us information about *only* that variable, irrespective of its relationship with others. The marginal PMF of $Y$ is simply the PMF of $Y$, ignoring $X$.

> **Theorem — Marginal PMF**
> $$f_Y(y) = \Pr[Y=y] = \sum_{x \in \text{Supp}[X]} f(x,y), \quad \forall y \in \mathbb{R}$$

The conditional distribution tells us the distribution of a variable given another has a specific value — what the PMF of $Y$ would be if we "threw out" all realizations except $X=x$.

> **Definition — Conditional PMF**
> $$f_{Y|X}(y|x) = \Pr[Y=y \mid X=x] = \frac{\Pr[X=x,Y=y]}{\Pr[X=x]} = \frac{f(x,y)}{f_X(x)}$$
> $\forall y \in \mathbb{R}$ and $\forall x \in \text{Supp}[X]$

> **Theorem — Multiplicative Law for PMFs**
> $$f_{X|Y}(x|y)f_Y(y) = f(x,y), \quad \forall x \in \mathbb{R}, \forall y \in \text{Supp}[Y]$$

## 5. Worked Example — Coin + Die

Flip a fair coin. Heads → roll a fair 4-sided die. Tails → roll a fair 6-sided die.

$$\Omega = \{(H,1),(H,2),(H,3),(H,4),(T,1),(T,2),(T,3),(T,4),(T,5),(T,6)\}$$

Let $A$ = heads, $B$ = observing 3. $P(A)=\tfrac12$, $P(B|A)=\tfrac14$. By the Multiplicative Law:

$$P(A\cap B) = P(B|A)P(A) = \frac14 \cdot \frac12 = \frac18 \qquad P(A^C\cap B) = \frac16 \cdot \frac12 = \frac{1}{12}$$

**Joint PMF** — $X=0$ (heads), $X=1$ (tails), $Y$ = die outcome:

$$
f(x,y) =
\begin{cases}
\frac18 & : x=0,\ y\in\{1,2,3,4\} \\
\frac{1}{12} & : x=1,\ y\in\{1,2,3,4,5,6\} \\
0 & : \text{otherwise}
\end{cases}
$$

**Joint CDF example:**

$$F(1,3) = \sum_{x\le1}\sum_{y\le3} f(x,y) = \frac18+\frac18+\frac18+\frac{1}{12}+\frac{1}{12}+\frac{1}{12} = \frac58$$

**Marginal PMFs:**

$$
f_X(x) =
\begin{cases}
\frac12 & : x=0 \\
\frac12 & : x=1
\end{cases}
\qquad
f_Y(y) =
\begin{cases}
\frac{5}{24} & : y\in\{1,2,3,4\} \\
\frac{1}{12} & : y\in\{5,6\}
\end{cases}
$$

**Why $P(Y=1)=\tfrac{5}{24}$?** Two paths:
- Heads → roll 1 on 4-sided die: $\tfrac12\times\tfrac14=\tfrac18$
- Tails → roll 1 on 6-sided die: $\tfrac12\times\tfrac16=\tfrac{1}{12}$
- Sum: $\tfrac18+\tfrac{1}{12}=\tfrac{5}{24}$ (same for 2,3,4)

**Why $P(Y=5)=\tfrac{1}{12}$?** Only via tails (heads die has no 5): $\tfrac12\times\tfrac16=\tfrac{1}{12}$ (same for 6).

**Conditional PMFs:**

$$
f_{X|Y}(x|y) =
\begin{cases}
\frac35 & : x=0,\ y\in\{1,2,3,4\} \\
\frac25 & : x=1,\ y\in\{1,2,3,4\} \\
1 & : x=1,\ y\in\{5,6\}
\end{cases}
\qquad
f_{Y|X}(y|x) =
\begin{cases}
\frac14 & : x=0,\ y\in\{1,2,3,4\} \\
\frac16 & : x=1,\ y\in\{1,...,6\}
\end{cases}
$$

> Marginal and conditional PMFs are still univariate PMFs — e.g. $f_{X|Y}(x|3)$ fixes $y=3$ and is a valid univariate PMF over $x$.

## 6. Jointly Continuous Random Variables

> **Definition — Jointly Continuous**
> $X,Y$ are jointly continuous if $\exists f: \mathbb{R}^2\to\mathbb{R}$, non-negative, such that
> $$F(x,y) = \int_{-\infty}^{x}\int_{-\infty}^{y} f(u,v)\,dv\,du, \quad \forall x,y \in \mathbb{R}$$

> **Definition — Joint PDF**
> $$f(x,y) = \left.\frac{\partial^2 F(u,v)}{\partial u\,\partial v}\right|_{u=x,v=y}$$

> **Theorem — Event Probabilities**
> For $D \subseteq \mathbb{R}^2$: $\quad \Pr[(X,Y)\in D] = \iint_D f(x,y)\,dy\,dx$

> **Theorem — Marginal PDF**
> $$f_Y(y) = \int_{-\infty}^{\infty} f(x,y)\,dx$$

> **Definition — Conditional PDF**
> $$f_{Y|X}(y|x) = \frac{f(x,y)}{f_X(x)}, \quad \forall y \in \mathbb{R}, \forall x \in \text{Supp}[X]$$

> **Theorem — Multiplicative Law for PDFs**
> $$f_{X|Y}(x|y)f_Y(y) = f(x,y)$$

## 7. Independence of Random Variables

> Marginals fully describe the distribution — **only under independence**.

Random variables can be statistically independent, since they're just real-valued functions of outcomes (same intuition as independence of events).

> **Definition — Independence of Random Variables**
> $X$ and $Y$ (discrete with joint PMF $f$, or jointly continuous with joint PDF $f$) are *independent* if $\forall x,y\in\mathbb{R}$:
> $$f(x,y) = f_X(x)f_Y(y)$$
> Written $X \perp Y$.

> 💡 Independence means the joint distribution contains no more information than the product of the marginals.

> **Theorem — Implications of Independence**
> The following are equivalent (each implies all others):
> - $X \perp Y$
> - $\forall x,y \in \mathbb{R},\ f(x,y) = f_X(x)f_Y(y)$
> - $\forall x \in \mathbb{R}, \forall y \in \text{Supp}[Y],\ f_{X|Y}(x|y) = f_X(x)$
> - $\forall D,E \subseteq \mathbb{R}$, events $\{X\in D\}$ and $\{Y\in E\}$ are independent
> - For all functions $g$ of $X$ and $h$ of $Y$: $g(X) \perp h(Y)$

---

## 🧾 Summary Cheat Sheet

**Joint** — probability of $X$ and $Y$ together: $f(x,y)$

**Marginal** — probability of one variable alone, summing/integrating out the other:
$$f_Y(y) = \sum_x f(x,y) \quad \text{or} \quad \int f(x,y)\,dx$$

**Conditional** — probability of $Y$ given $X=x$:
$$f_{Y|X}(y|x) = \frac{f(x,y)}{f_X(x)}$$

**Independence** — joint factors into marginals:
$$f(x,y) = f_X(x)f_Y(y)$$

**Key limitation** — marginals alone never determine the joint (unless independent).
