## Joint Probability

For $A, B \in S$, the **joint probability** of $A$ and $B$ is $P(A \cap B)$.

---

## Mutually Exclusive Events

### $(A \cap B) = \emptyset$

Two events are **mutually exclusive** (disjoint) if they cannot happen at the same time in the same experiment — they are impossible together.

**Example:** Rolling a die — the events "rolling a 4" and "rolling a 6" are mutually exclusive because only one outcome can occur in a single roll.

> ⚠️ **Important:** The word "AND" in probability does **not** automatically mean multiplying probabilities. "AND" means the **intersection** of events — the outcomes satisfying both conditions simultaneously. We only multiply probabilities when dealing with independent experiments or independent events.

**Example — one die roll:**

Let $A = \{4, 5, 6\}$ (greater than or equal to 4) and $B = \{2, 4, 6\}$ (even).

"A AND B" means outcomes that are both even **and** ≥ 4, giving $\{4, 6\}$.

Since a die can only land on one face at a time, these outcomes are mutually exclusive inside the intersection, so we **add** their probabilities:

$$P(A \cap B) = \frac{1}{6} + \frac{1}{6} = \frac{1}{3}$$

Multiplication is used when "AND" refers to **separate independent experiments** — like flipping a coin AND rolling a die.

## Addition Rule — Two Events

<p align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/addition_rule_2_case_1.1.png" width="500">
</p>

For $A, B \in S$:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

**Proof.** Note that $(A \setminus B)$, $(B \setminus A)$, and $(A \cap B)$ are pairwise disjoint and

$(A \cup B) = (A \setminus B) \cup (B \setminus A) \cup (A \cap B)$

so by countable additivity,

$P(A \cup B) = P(A \setminus B) + P(B \setminus A) + P(A \cap B)$

$= P(A \setminus (A \cap B)) + P(B \setminus (A \cap B)) + P(A \cap B)$

$= P(A) - P(A \cap B) + P(B) - P(A \cap B) + P(A \cap B)$

$= P(A) + P(B) - P(A \cap B)$

where the second equality holds because

$A \setminus B = \emptyset \cup (A \setminus B)$

$= (A \cap A^C) \cup (A \cap B^C)$

$= A \cap (A^C \cup B^C)$

$= A \cap (A \cap B)^C$

$= A \setminus (A \cap B)$

and likewise $B \setminus A = B \setminus (A \cap B)$, while the third equality follows from the subtraction rule, since $A \cap B \subseteq A$ and $A \cap B \subseteq B$. $\square$

## Addition Rule — Three Events

For any three events $A$, $B$, $C$ in a probability space:

$P(A \cup B \cup C) = P(A) + P(B) + P(C) - P(A \cap B) - P(A \cap C) - P(B \cap C) + P(A \cap B \cap C)$

**Proof.** Let $D = B \cup C$. Then by the addition rule for two events:

$P(A \cup B \cup C) = P(A \cup D) = P(A) + P(D) - P(A \cap D)$

**Expanding $P(D)$:** applying the two-event rule to $B \cup C$:

$P(D) = P(B \cup C) = P(B) + P(C) - P(B \cap C)$

**Expanding $P(A \cap D)$:** using the distributive law $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$, then the two-event rule:

$P(A \cap D) = P((A \cap B) \cup (A \cap C)) = P(A \cap B) + P(A \cap C) - P(A \cap B \cap C)$

**Substituting back:**

$P(A \cup B \cup C) = P(A) + P(B) + P(C) - P(B \cap C) - P(A \cap B) - P(A \cap C) + P(A \cap B \cap C)$

$= P(A) + P(B) + P(C) - P(A \cap B) - P(A \cap C) - P(B \cap C) + P(A \cap B \cap C) \quad \square$

</p>

## Conditional Probability

Conditional probability means that instead of looking at the entire population, we first **restrict or filter our world** to only the cases where some condition is true, and then measure probability inside that smaller world. It is not necessarily "effect given cause" — it is simply:

> *"The probability of one event assuming another event already happened."*

**Example:**

$$P(\text{Lung Cancer} \mid \text{Smoker})$$

We are asking: *"Among smokers only, what fraction have lung cancer?"*

Suppose we have 1000 people total, and among them 200 are smokers. Out of those 200 smokers, 50 have lung cancer.

In conditional probability, we no longer use the whole 1000 people as our world. We **shrink our sample space** to only the smokers:

- World before: 1000 people
- World after conditioning: 200 smokers only
- Lung cancer cases inside that world: 50

$$P(\text{Lung Cancer} \mid \text{Smoker}) = \frac{50}{200} = 0.25$$

> **Core intuition:** Conditional probability changes the **denominator** by filtering the world to only the condition we are given.

### Definition — *Conditional Probability*

For $A, B \in S$ with $P(B) > 0$, the *conditional probability* of $A$ given $B$ is

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

---

## Multiplicative Law of Probability

### Theorem — *Multiplicative Law of Probability*

For $A, B \in S$ with $P(B) > 0$:

$$P(A \mid B)\, P(B) = P(A \cap B)$$

## Bayes' Rule

Bayes' Rule relates the conditional probability of $A$ given $B$ to the conditional probability of $B$ given $A$. It is most often used for **reasoning backward from evidence to hidden causes**.

Imagine a doctor trying to understand whether a patient is likely to be a smoker after seeing that the patient has lung cancer. Normally, it is easier to measure the forward direction: *"Among smokers, what fraction develop lung cancer?"* because we can observe many smokers over time. But the doctor's real question is the reverse: *"Among people who already have lung cancer, what fraction are smokers?"*

Bayes' theorem helps us reverse this reasoning — instead of starting from the possible cause (smoking) and looking for the effect (lung cancer), Bayes starts from the observed evidence (lung cancer) and updates our belief about the hidden cause (smoking).

> **Core idea:** When new evidence appears, we shrink our world to only the people with that evidence, and then re-evaluate how likely each possible cause is inside that new filtered world.

---

### The Easy Direction — Forward

$$P(\text{Lung Cancer} \mid \text{Smoker})$$

*"Among smokers, what fraction get lung cancer?"*

This is relatively straightforward — we begin with smokers and observe outcomes over time.

---

### The Hard Direction — Backward

$$P(\text{Smoker} \mid \text{Lung Cancer})$$

*"Among people who already have lung cancer, what fraction are smokers?"*

Now we are reasoning **backward** from evidence (lung cancer) toward a possible cause (smoking). Bayes' Rule connects these two worlds by using the easier forward probability to estimate the harder reverse probability.

> **Core intuition:** Conditional probability alone is simply about filtering the sample space. Bayes becomes powerful because it lets us **update our belief about hidden causes after seeing evidence.**

---

### Definition — Bayes' Rule

For $A, B \in S$ with $P(B) > 0$:

$$\boxed{P(A \mid B) = \frac{P(B \mid A)\, P(A)}{P(B)}}$$

---

### Bayes' Rule — Cause and Effect Form

Let $C$ denote a **cause** and $E$ denote observed **evidence**. Then:

$$P(\text{Cause} \mid \text{Effect}) = \frac{P(\text{Effect} \mid \text{Cause})\, P(\text{Cause})}{P(\text{Effect})}$$

$$P(\text{Hypothesis} \mid \text{Evidence}) = \frac{P(\text{Evidence} \mid \text{Hypothesis})\, P(\text{Hypothesis})}{P(\text{Evidence})}$$

$$\text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Normalization}}$$

| Term | Name | Meaning |
|------|------|---------|
| P(Cause given Effect) | Posterior | Probability of the cause after seeing the evidence |
| P(Effect given Cause) | Likelihood | How probable is the evidence if the cause is true? |
| P(Cause) | Prior | Initial belief in the cause before seeing evidence |
| P(Effect) | Marginal | Total probability of the evidence under all causes |

> **In one line:** Bayes' Rule tells you how to rationally update your belief about a hidden cause once you observe new evidence.

## Frequentist vs. Bayesian Thinking

| | Frequentist | Bayesian |
|---|---|---|
| What is probability? | Long-run frequency of repeated experiments | Degree of belief given current evidence |
| Is the unknown fixed? | Yes — fixed fact, even if unknown | No — uncertain, described by a probability |
| How does uncertainty update? | It does not — only repeated experiments matter | It updates when new evidence arrives |

A **frequentist** thinks probability is how often something happens in repeated experiments over the long run. A **Bayesian** thinks probability is how strongly we believe something is true given current evidence.

**Frequentist view:** The patient either has the disease or does not — it is a fixed fact, even if unknown. Probability only comes from repeating the same experiment many times. If a test is "95% accurate," a frequentist interprets this as: if we repeat this test on thousands of patients who truly have the disease, the test will correctly detect it about 95% of the time. Uncertainty is about the long-run behavior of the testing process, not about the individual patient.

**Bayesian view:** Before seeing any test result, a doctor might believe there is only a 10% chance a patient has a disease based on age, symptoms, or population statistics — this is the **prior belief**. After a positive test result, the doctor updates that belief using the new evidence and may now believe there is an 80% chance the patient has the disease — this is the **posterior belief**.

Unlike the frequentist view, Bayesians treat probability as something that can **change when new information arrives**. The entire process of starting with a prior belief, observing evidence, and updating to a posterior belief is exactly what Bayes' theorem formalizes.

## Partition

### Definition — *Partition*

<p align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/partition.png" width="500">
</p>

If $A_1, A_2, A_3, \ldots \in S$ are **nonempty** and **pairwise disjoint**, and $\Omega = A_1 \cup A_2 \cup A_3 \cup \ldots$, then $\{A_1, A_2, A_3, \ldots\}$ is a *partition* of $\Omega$.

> A partition divides the sample space into **mutually exclusive** and **exhaustive** categories — like bins that together hold everything and never overlap.

---

## Law of Total Probability

### Theorem — *Law of Total Probability*

The law of total probability says that the probability of an event can be found by **breaking it into all possible mutually exclusive cases** and adding their probabilities.

<p align="center">
  <img src="https://raw.githubusercontent.com/alilordifar/ucb-stat-203-notes/master/images/total_probability.png" width="500">
</p>

If $\{A_1, A_2, A_3, \ldots\}$ is a partition of $\Omega$ and $B \in S$, then

$$P(B) = \sum_i P(B \cap A_i)$$

If we also have $P(A_i) > 0$ for $i = 1, 2, 3, \ldots$, then this can also be stated as

$$P(B) = \sum_i P(B \mid A_i)\, P(A_i)$$

---

## Alternative Forms of Bayes' Rule

### Theorem 1.1.14 — *Alternative Forms of Bayes' Rule*

If $\{A_1, A_2, A_3, \ldots\}$ is a partition of $\Omega$ with $P(A_i) > 0$ for $i = 1, 2, 3, \ldots$, and $B \in S$ with $P(B) > 0$, then

$$P(A_j \mid B) = \frac{P(B \mid A_j)\, P(A_j)}{\sum_i P(B \cap A_i)}$$

or equivalently,

$$P(A_j \mid B) = \frac{P(B \mid A_j)\, P(A_j)}{\sum_i P(B \mid A_i)\, P(A_i)}$$

---

## Independence

Independence of events means **knowing one thing gives you no information about the other**. Two events $A$ and $B$ are independent if whether $A$ happens does not change the probability that $B$ happens.

### Definition — *Independence of Events*

Events $A, B \in S$ are *independent* if

$$P(A \cap B) = P(A)\, P(B)$$

---

### Theorem — *Conditional Probability and Independence*

For $A, B \in S$ with $P(B) > 0$, $A$ and $B$ are independent if and only if

$$P(A \mid B) = P(A)$$

> Loosely speaking, when $A$ and $B$ are independent, knowing whether or not $B$ has occurred gives us **no information** about the probability that $A$ has occurred.
