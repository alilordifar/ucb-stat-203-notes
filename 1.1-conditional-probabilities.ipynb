# Joint and Conditional Probability

For $A, B \in S$ the joint probability of $A$ and $B$ is $P(A\cap B)$ 

### Mutual Exclusive Events (Disjoint Events) -> $(A\cap B) = \emptyset$
Two events are mutually exclusive if they cannot happen at the same time in the same experiment. In other words, they are impossible together. For example, in rolling a die, the events “rolling a 4” and “rolling a 6” are mutually exclusive because only one outcome can occur in a single roll. Also, the word “AND” in probability does not always mean multiplying probabilities together. “AND” means the intersection of events — the outcomes satisfying both conditions simultaneously. We only multiply probabilities when dealing with independent experiments or independent events.

When we say “A AND B” in probability, we do not automatically multiply probabilities; instead, we first find the outcomes that satisfy both conditions at the same time (the intersection). For one die roll, if A={4,5,6} (“greater than or equal to 4”) and B={2,4,6} (“even”), then “A AND B” means outcomes that are both even and ≥4, which gives {4,6}. Now inside this resulting event, either 4 can happen or 6 can happen, and since a die can only land on one face at a time, these outcomes are mutually exclusive, so we add their probabilities: 1/6 + 1/6 = 1/3
. Multiplication is used when “AND” refers to separate independent experiments, like flipping a coin AND rolling a die.

### Addition Rule
For $A, B \in S$,  $P(A\cup B) = P(A) + P(B) - P(A\cap B)$ 

**Proof.** Note that $(A \setminus B)$, $(B \setminus A)$, and $(A \cap B)$ are pairwise disjoint and

$(A \cup B) = (A \setminus B) \cup (B \setminus A) \cup (A \cap B),$

so by countable additivity,

$P(A \cup B) = P(A \setminus B) + P(B \setminus A) + P(A \cap B)$

$= P(A \setminus (A \cap B)) + P(B \setminus (A \cap B)) + P(A \cap B)$

$= P(A) - P(A \cap B) + P(B) - P(A \cap B) + P(A \cap B)$

$= P(A) + P(B) - P(A \cap B),$

where the second equality holds because

$A \setminus B = \emptyset \cup (A \setminus B)$

$= (A \cap A^C) \cup (A \cap B^C)$

$= A \cap (A^C \cup B^C)$

$= A \cap (A \cap B)^C$

$= A \setminus (A \cap B),$

and likewise $B \setminus A = B \setminus (A \cap B)$, while the third equality follows from the
subtraction rule, since $A \cap B \subseteq A$ and $A \cap B \subseteq B$. $\square$

### Theorem — Addition Rule for Three Events

For any three events $A$, $B$, $C$ in a probability space:

$P(A \cup B \cup C) = P(A) + P(B) + P(C) - P(A \cap B) - P(A \cap C) - P(B \cap C) + P(A \cap B \cap C)$

**Proof.** Let $D = B \cup C$. Then by the addition rule for two events:

$P(A \cup B \cup C) = P(A \cup D) = P(A) + P(D) - P(A \cap D)$

**Expanding $P(D)$:** applying the two-event rule to $B \cup C$:

$P(D) = P(B \cup C) = P(B) + P(C) - P(B \cap C)$

**Expanding $P(A \cap D)$:** using the distributive law $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$, then the two-event rule:

$P(A \cap D) = P((A \cap B) \cup (A \cap C)) = P(A \cap B) + P(A \cap C) - P(A \cap B \cap C)$

**Substituting back:**

$P(A \cup B \cup C) = P(A) + P(B) + P(C) - P(B \cap C) - P(A \cap B) - P(A \cap C) + P(A \cap B \cap C)$

$= P(A) + P(B) + P(C) - P(A \cap B) - P(A \cap C) - P(B \cap C) + P(A \cap B \cap C) \quad \square$
