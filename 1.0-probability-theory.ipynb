## Random Phenomenon

> A process that satisfies all three conditions:

1. Can be **repeated under identical conditions**
2. Has outcomes that **cannot be predicted** with certainty in advance
3. Produces outcomes from a **well-defined set**

---

## Random Experiment

> A procedure with a well-defined set of all possible outcomes, where the outcome of every single trial is **uncertain**.

---

## Random Experiment vs. Random Phenomenon

| | Definition | Example |
|---|---|---|
| **Random Phenomenon** | The uncertain process in nature/reality | "Uncertainty in tomorrow's weather" |
| **Random Experiment** | The action/procedure used to observe outcomes | "Measure tomorrow's temperature/rain" |

---

## Random Generative Process

> A mechanism that **selects one outcome** from among multiple possible outcomes each time it runs.

- It is **repeatable** — the mechanism can run many times
- Each run independently selects exactly one outcome

**Example:** A coin flip is the simplest random generative process — each flip selects either heads or tails.

---

## Outcome

> An **outcome** (also called an *elementary*, *atomic*, or *simple* event) is a single element of the sample space — one result that actually happens on a given trial.

---

## Sample Space

> The **sample space** $\Omega$ is the set of **all possible outcomes** of the random generative process.

Individual outcomes (also called *sample points*) are denoted:

$$\omega \in \Omega$$

---

## Event Space

> **Events** are subsets of $\Omega$. The **event space** is the collection of all allowed events.

**Example:** Rolling a single six-sided die — the event of rolling an even number is:

$$A = \{\omega \in \Omega : \omega \text{ is even}\} = \{2, 4, 6\}$$

> **Note:** An event can also correspond to a single state of the world — for instance, the event of rolling a 3, represented by $B = \{3\}$. Such events are variously known as *atomic events*, *elementary events*, or *simple events*.

### Definition — *Event Space*

A set $S$ of subsets of $\Omega$ is an **event space** if it satisfies the following:

- **Nonempty:** $S \neq \emptyset$
- **Closed under complements:** if $A \in S$, then $A^C \in S$
- **Closed under countable unions:** if $A_1, A_2, A_3, \ldots \in S$, then $A_1 \cup A_2 \cup A_3 \cup \ldots \in S$

### Power Set

> The set of **all possible subsets** of a set, including the empty set and the set itself. If a set has $n$ elements, its power set has $2^n$ subsets.

---

## Sample Space vs. Event Space

The **sample space** $\Omega$ is the set of all individual outcomes a random process can produce, while the **event space** $\mathcal{F}$ is the collection of groups of outcomes (events) that we can assign probabilities to.

**Example — rolling a die:**

| Concept | Content |
|---------|---------|
| Sample space $\Omega$ | $\{1, 2, 3, 4, 5, 6\}$ — the actual possible outcomes |
| Event | $\{2, 4, 6\}$ — "getting an even number" |
| Event space $\mathcal{F}$ | Contains $\{2,4,6\}$, $\{1,3,5\}$, $\{6\}$, $\emptyset$, $\Omega$, ... |

> **Sample space** = raw possible outcomes. **Event space** = allowed subsets/groups of those outcomes.

> **Note:** An event is *not* one outcome — it is a **set** of outcomes. An event is a condition that becomes true if certain outcomes happen, so mathematically it is represented as a set of outcomes. It can contain one outcome (like "rolling a 3" $= \{3\}$) or multiple outcomes (like "rolling an even number" $= \{2,4,6\}$). The event space is simply the collection of all events we are allowed to talk about and assign probabilities to.

## Sigma Algebra & Event Space

> **Short answer:** In formal probability theory, event space = sigma algebra. They are the same thing.

### The Probability Triple

$$(\Omega,\ \mathcal{F},\ P)$$

| Symbol | Name | Meaning |
|--------|------|---------|
| $\Omega$ | Sample space | All possible raw outcomes |
| $\mathcal{F}$ | Event space = Sigma algebra | All valid *groups* of outcomes you can assign probabilities to |
| $P$ | Probability measure | Function that assigns a number to each event in $\mathcal{F}$ |

### Example — Six-sided die

**Sample space:**
$$\Omega = \{1, 2, 3, 4, 5, 6\}$$

**A valid (small) sigma algebra:**
$$\mathcal{F} = \{\emptyset,\ \{1,3,5\},\ \{2,4,6\},\ \Omega\}$$

This lets us assign probabilities to: odd, even, impossible ($\emptyset$), and everything ($\Omega$).

### Power Set vs. Sigma Algebra

For $\Omega = \{1,2,3,4,5,6\}$, the **power set** (largest possible sigma algebra) has $2^6 = 64$ subsets.

| Concept | Meaning |
|---------|---------|
| Power set | **All** possible subsets of $\Omega$ |
| Sigma algebra | **Some** approved subsets satisfying the probability rules |
| Largest sigma algebra | The power set itself |

> In **beginner** probability: $\mathcal{F}$ is usually the full power set.
> In **advanced** probability: smaller sigma algebras are sometimes used intentionally.

The event space (sigma algebra) is a collection of subsets of the sample space that we are allowed to assign probabilities to. In advanced probability, we may use a **smaller sigma algebra** containing only the events we care about, but it must still satisfy sigma algebra rules — so it cannot contain just $\{2,4,6\}$ alone.

For a die, if we only care about "even," the smallest valid sigma algebra is:

$$\mathcal{F} = \{\emptyset,\ \{2,4,6\},\ \{1,3,5\},\ \Omega\}$$

because it must also include the **complement** (odd numbers), the **empty set**, and the **full sample space**.

### Why "Sigma Algebra" and not just "Event Space"?

Because the collection must satisfy specific rules for probability to work consistently:

- **Nonempty:** $\mathcal{F} \neq \emptyset$
- **Closed under complements:** if $A \in \mathcal{F}$, then $A^C \in \mathcal{F}$
- **Closed under countable unions:** if $A_1, A_2, \ldots \in \mathcal{F}$, then $A_1 \cup A_2 \cup \ldots \in \mathcal{F}$

The small example above satisfies all three: complement of odd is even, complement of even is odd, and unions stay inside the collection.

## Frequentist Probability

> **Frequentism** interprets probability as the **long-run proportion** of times an event occurs across many repetitions/realizations of a random generative process.

Formally, the probability of an event $A$ represents how frequently $A$ would occur among many, many draws:

$$P(A) = \lim_{n \to \infty} \frac{\text{number of times } A \text{ occurs}}{n}$$

| Concept | Meaning |
|---------|---------|
| $A$ | The event of interest |
| $n$ | Total number of trials |
| $P(A)$ | Long-run frequency of $A$ |

> 💡 Under frequentism, probability only makes sense for **repeatable** processes — it describes what *would* happen over many realizations, not any single trial.

---

## Probability Measure

> A **probability measure** is a function $P: S \to \mathbb{R}$ that assigns a probability to every event in the event space.

---

## Kolmogorov Axioms

Before Andrey Kolmogorov, probability was used successfully, but its foundations were somewhat informal and interpreted through ideas like *frequentist probability, gambling intuition, geometric probability, subjective reasoning, etc.* People could compute probabilities, but there was no single rigorous mathematical foundation tying everything together consistently.

Kolmogorov's major contribution in 1933 was defining probability **axiomatically** using set theory and measure theory. He formalized probability as $(\Omega, \mathcal{F}, P)$.

> Frequentists tried to explain what probability *means* philosophically. Kolmogorov defined how probability must *mathematically behave* — regardless of interpretation.

### Definition — *Kolmogorov Axioms*

Let $\Omega$ be a sample space, $S$ be an event space, and $P$ be a probability measure.
Then $(\Omega, S, P)$ is a **probability space** if it satisfies the following:

- **Non-negativity:** $\forall A \in S,\ P(A) \geq 0$, where $P(A)$ is finite and real.
- **Unitarity:** $P(\Omega) = 1$
- **Countable additivity:** if $A_1, A_2, A_3, \ldots \in S$ are pairwise disjoint, then

$$P(A_1 \cup A_2 \cup A_3 \cup \ldots) = P(A_1) + P(A_2) + P(A_3) + \ldots = \sum_i P(A_i)$$

**Interpretation of each axiom:**

| Axiom | Meaning |
|-------|---------|
| **Non-negativity** | No event can have a negative probability |
| **Unitarity** | It is certain that *some* outcome will occur |
| **Countable additivity** | For mutually exclusive events, probabilities add up |

## Basic Properties of Probability

### Theorem — *Basic Properties of Probability*

Let $(\Omega, S, P)$ be a probability space. Then:

- **Monotonicity:** $\forall A, B \in S$, if $A \subseteq B$, then $P(A) \leq P(B)$
- **Subtraction rule:** $\forall A, B \in S$, if $A \subseteq B$, then $P(B \setminus A) = P(B) - P(A)$
- **Zero probability of the empty set:** $P(\emptyset) = 0$
- **Probability bounds:** $\forall A \in S,\ 0 \leq P(A) \leq 1$
- **Complement rule:** $\forall A \in S,\ P(A^C) = 1 - P(A)$

---

### Proof

Let $A, B \in S$ with $A \subseteq B$. Since $B = A \cup (B \setminus A)$, and $A$ and $(B \setminus A)$ are disjoint, countable additivity implies:

$$P(B) = P(A) + P(B \setminus A)$$

Rearranging, non-negative probabilities then imply monotonicity: $P(B \setminus A) \geq 0$, so

$$P(A) = P(B) - P(B \setminus A) \leq P(B)$$

Rearranging again yields the subtraction rule:

$$P(B \setminus A) = P(B) - P(A)$$

The subtraction rule implies zero probability of the empty set: $A \subseteq A$, so

$$P(\emptyset) = P(A \setminus A) = P(A) - P(A) = 0$$

Monotonicity and unitarity (and non-negativity) imply the probability bounds: since $A \subseteq \Omega$,

$$0 \leq P(A) \leq P(\Omega) = 1$$

Finally, the subtraction rule and unitarity imply the complement rule:

$$P(A^C) = P(\Omega \setminus A) = P(\Omega) - P(A) = 1 - P(A) \qquad \square$$

**Intuition behind each property:**

- **Monotonicity** — if one event is a subset of another (so the former always occurs whenever the latter does), then the probability of the former is no greater than that of the latter.
- **Subtraction rule** — the probability that the second event occurs but not the first equals the probability of the second minus the probability of the first.
- **Zero probability of the empty set** — some event in our event space must always occur.
- **Probability bounds** — each event has a probability between zero and one.
- **Complement rule** — the probability of an event *not* occurring is one minus the probability of it occurring, so the probability that a given event either occurs or does not occur is always one.
